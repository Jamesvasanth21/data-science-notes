## 1. Rule-Based Methods

**Definition:**  
Rule-based methods use predefined linguistic rules, patterns, and dictionaries to process text. These methods are deterministic and are best suited for tasks where domain knowledge is critical.

**Examples:**
- Tokenization using regular expressions
- Part-of-speech tagging using handcrafted rules
- Grammar checking tools



**Example: Simple Regex-based Tokenizer**

```python
import re

text = "Natural Language Processing is fascinating!"

# Split on spaces and punctuation
tokens = re.findall(r'\b\w+\b', text.lower())

print(tokens)
```

## 2. Machine Learning Approaches

These approaches learn patterns from data instead of using predefined rules.

### ✅ Naive Bayes for NLP

**Use Case:**
Text classification, spam detection.

**Key Idea:**
Uses probabilities based on word occurrences assuming feature independence.


**Example: Spam Detection**

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Sample data
texts = ["Congratulations! You've won a free ticket.",
         "Meeting at 10am today.",
         "Claim your prize now!",
         "Let's catch up tomorrow."]

labels = [1, 0, 1, 0]  # 1: spam, 0: not spam

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

model = MultinomialNB()
model.fit(X, labels)

# Predict on new message
new_text = ["Free lottery winner!"]
X_new = vectorizer.transform(new_text)
prediction = model.predict(X_new)

print("Spam" if prediction[0] == 1 else "Not Spam")
```

## 3. Deep Learning Models

Deep learning models use neural networks to automatically learn patterns from text data.

### ✅ Recurrent Neural Networks (RNN)

**Use Case:**
Sequence modeling tasks like text generation and language modeling.

**Key Idea:**
Processes sequences step-by-step, maintaining a hidden state across time steps.

**Example:**
[TensorFlow RNN Tutorial](https://www.tensorflow.org/text/tutorials/text_generation)

---

### ✅ Long Short-Term Memory (LSTM)

**Use Case:**
Text summarization, machine translation.

**Key Idea:**
Addresses the vanishing gradient problem in RNNs using gates to control information flow.

**Example:**
[Keras LSTM Text Classification](https://keras.io/examples/nlp/text_classification_with_lstm/)

---

### ✅ Convolutional Neural Networks (CNN) for NLP

**Use Case:**
Sentence classification, document analysis.

**Key Idea:**
Applies convolutional filters to capture local patterns like phrases or n-grams.

**Example:**
[Text classification using CNN](https://github.com/keras-team/keras/blob/master/examples/imdb_cnn_lstm.py)

---

### ✅ Transformer Models (BERT, GPT)

**Use Case:**
Contextual understanding, language generation, question answering.

**Key Idea:**
Uses self-attention mechanisms to model relationships between all words in a sequence.

**Popular Models:**

* **BERT (Bidirectional Encoder Representations from Transformers):** Pretrained on masked language modeling for bidirectional context understanding.
* **GPT (Generative Pretrained Transformer):** Uses autoregressive modeling for text generation.


**Example:**

```python
from transformers import pipeline

qa_pipeline = pipeline("question-answering")

context = "BERT is a transformer-based model used in NLP."
question = "What is BERT used for?"

result = qa_pipeline(question=question, context=context)
print(result)
```
