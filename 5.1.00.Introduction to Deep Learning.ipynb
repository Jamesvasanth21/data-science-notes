### 🔹 **Neural Networks**

* **Definition:**
  A **neural network** is a computational model inspired by the human brain. It consists of **layers of interconnected nodes (neurons)** that transform inputs to outputs through weights, biases, and activation functions.

* **Types:**

  * **Perceptron** (single-layer)
  * **Shallow Neural Networks** (1 hidden layer)
  * **Deep Neural Networks** (2+ hidden layers)

* **Key Components:**

  * Input layer
  * Hidden layer(s)
  * Output layer
  * Weights, biases
  * Activation functions

* **Use Cases:**
  Basic regression, classification, pattern recognition, etc.

### 🔹 **Deep Learning**

* **Definition:**
  Deep Learning is a **subset of machine learning** that uses **neural networks with many (deep) layers**, typically more than two hidden layers. These deep models can **automatically learn complex features** from large amounts of data.

* **Key Characteristics:**

  * Involves **Deep Neural Networks (DNNs)**
  * Often uses architectures like:

    * **RNNs/LSTMs** – sequences/time series
    * **Transformers** – language, vision, etc.

* **Use Cases:**
  Image classification, natural language processing, speech recognition, self-driving cars, etc.

# 🧠 **Neural Networks Basics**

Neural Networks (NNs) are the building blocks of deep learning. Inspired by the human brain, they consist of interconnected layers of units called **neurons**.

Deep Learning is a powerful subfield of Machine Learning that uses **Artificial Neural Networks (ANNs)**, inspired by the structure and function of the human brain, to learn complex patterns from data. These networks are "deep" because they consist of multiple layers of interconnected nodes (neurons).

The core idea is to train these networks on vast amounts of data to automatically extract features and make predictions or classifications without explicit programming for specific tasks. This ability to learn hierarchical representations of data makes deep learning exceptionally effective for complex tasks like image recognition, natural language processing, and speech recognition.

## 1\. Neurons and Connections

Each connection between neurons has an associated **weight ($w$)**, and each neuron has a **bias ($b$)**. When a signal passes through a connection, it's multiplied by the weight. The weighted sum of inputs, plus the bias, is then passed through an **activation function** to produce the neuron's output.

**Formula for a single neuron:**
Let $\mathbf{x} = [x_1, x_2, \ldots, x_n]$ be the inputs from the previous layer, and $\mathbf{w} = [w_1, w_2, \ldots, w_n]$ be the weights for these inputs, and $b$ be the bias.

1.  **Weighted Sum (Linear Transformation):**
    $$z = (\mathbf{w} \cdot \mathbf{x}) + b = \sum_{i=1}^{n} w_i x_i + b$$
2.  **Activation Function:**
    $$a = f(z)$$
    Where $f$ is the activation function.

## 2\. Artificial Neural Networks (ANNs) Layers

An ANN is composed of layers of interconnected "neurons" or "nodes."

  * **Input Layer**: Receives the raw data. Each node typically corresponds to a feature in your dataset.
  * **Hidden Layers**: One or more layers between the input and output layers. These layers perform non-linear transformations on the input data, extracting increasingly abstract and complex features. The "depth" of a neural network refers to the number of hidden layers.
  * **Output Layer**: Produces the final prediction or classification. The number of nodes here depends on the problem (e.g., 1 for regression, multiple for multi-class classification).

## 3\. Activation Functions

Activation functions introduce **non-linearity** into the network, allowing it to learn complex, non-linear relationships in the data. Without them, even a deep network would behave like a simple linear model.

Common activation functions:

  * **ReLU (Rectified Linear Unit)**: $f(z) = \max(0, z)$
      * Widely used in hidden layers due to its simplicity and effectiveness in mitigating vanishing gradients.
  * **Sigmoid**: $f(z) = \frac{1}{1 + e^{-z}}$
      * Squashes values between 0 and 1, often used in the output layer for binary classification (representing probabilities).
  * **Tanh (Hyperbolic Tangent)**: $f(z) = \frac{e^z - e^{-z}}{e^z + e^{-z}}$
      * Squashes values between -1 and 1.
  * **Softmax**: $f(\mathbf{z})*j = \frac{e^{z\_j}}{\\sum*{k=1}^{K} e^{z\_k}}$
      * Used in the output layer for multi-class classification, converting a vector of raw scores into a probability distribution where all probabilities sum to 1.

A **simple reference table** for common activation functions in neural networks:

| Activation Function | Formula                                                            | Output Range         | Common Use Case                       |
| ------------------- | ------------------------------------------------------------------ | -------------------- | ------------------------------------- |
| **Sigmoid**         | $\sigma(x) = \frac{1}{1 + e^{-x}}$                                 | $(0, 1)$             | Binary classification (probabilities) |
| **Tanh**            | $\tanh(x) = \frac{e^{x} - e^{-x}}{e^{x} + e^{-x}}$                 | $(-1, 1)$            | Hidden layers for normalized outputs  |
| **ReLU**            | $f(x) = \max(0, x)$                                                | $[0, \infty)$        | Most common for hidden layers         |
| **Leaky ReLU**      | $f(x) = \begin{cases} x & x \ge 0 \\ \alpha x & x < 0 \end{cases}$ | $(-\infty, \infty)$  | Avoids dying ReLU problem             |
| **Softmax**         | $f(x_i) = \frac{e^{x_i}}{\sum_{j} e^{x_j}}$                        | $(0, 1)$ (sums to 1) | Multi-class classification            |
| **Linear**          | $f(x) = x$                                                         | $(-\infty, \infty)$  | Regression outputs                    |

## 4\. Loss Function (Cost Function)

A loss function quantifies the **error** between the model's predicted output and the actual target output. The goal during training is to minimize this loss.

  * **Regression**:
      * **Mean Squared Error (MSE)**:
$L = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$

      * **Mean Absolute Error (MAE)**:
$L = \frac{1}{N} \sum_{i=1}^{N} |y_i - \hat{y}_i|$

  * **Classification**:
      * **Binary Cross-Entropy**: $L = -\frac{1}{N} \sum_{i=1}^{N} [y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i)]$ (for binary classification)

      * **Categorical Cross-Entropy**: $L = -\frac{1}{N} \sum_{i=1}^{N} \sum_{j=1}^{K} y_{ij} \log(\hat{y}_{ij})$ (for multi-class classification)  



## 5\. Optimization Algorithm: Gradient Descent

Gradient Descent is an iterative optimization algorithm used to find the set of weights and biases that minimize the loss function.

  * It calculates the **gradient** (the partial derivatives of the loss function with respect to each weight and bias). The gradient points in the direction of the steepest increase in loss.
  * The weights and biases are then updated in the **opposite direction** of the gradient, scaled by a **learning rate ($\\alpha$)**, to move towards the minimum loss.

**Update Rule:**
$$W_{new} = W_{old} - \alpha \frac{\partial L}{\partial W}$$
$$b_{new} = b_{old} - \alpha \frac{\partial L}{\partial b}$$

Different variants exist:

  * **Batch Gradient Descent**: Uses the entire dataset to compute the gradient for each update. Slow for large datasets.
  * **Stochastic Gradient Descent (SGD)**: Updates weights for each individual training example. Faster but can be noisy.
  * **Mini-Batch Gradient Descent**: Updates weights for small batches of training examples. A good compromise between speed and stability, commonly used.


**100 rows of data**:

---

**1️⃣ Batch size = 1 (Stochastic Gradient Descent)**

* **Process:** Model updates weights **after every single sample**.
* **Per epoch:**

  * 100 updates (because 100 samples)
  * Sequence: take 1 row → forward pass → loss → backward pass → weight update → repeat for next row
* **Advantages:**

  * More frequent updates → can escape local minima.
  * Good for small datasets.
* **Disadvantages:**

  * Training is slower per epoch.
  * More noise in updates.

---

**2️⃣ Batch size = 50 (Mini-Batch Gradient Descent)**

* **Process:** Model updates weights **after every 50 samples**.
* **Per epoch:**

  * 100 samples ÷ 50 = **2 updates** per epoch.
  * Sequence: take 50 rows → forward pass on all 50 → average loss → backward pass → weight update.
* **Advantages:**

  * Balance between speed and stability.
  * Uses vectorized operations efficiently.
* **Disadvantages:**

  * May require tuning to find optimal batch size.

---

**3️⃣ Batch size = 100 (Batch Gradient Descent)**

* **Process:** Model updates weights **once per epoch**, using **all 100 samples** together.
* **Per epoch:**

  * 1 update (all 100 samples at once).
* **Advantages:**

  * Very stable updates.
  * Works well if data fits in memory.
* **Disadvantages:**

  * May converge slowly.
  * Can get stuck in local minima.

---

| Batch Size | Updates per Epoch | Type of Gradient Descent | Speed  | Stability | Noise in Update |
| ---------- | ----------------- | ------------------------ | ------ | --------- | --------------- |
| **1**      | 100               | Stochastic               | Slow   | Low       | High            |
| **50**     | 2                 | Mini-Batch               | Medium | Medium    | Medium          |
| **100**    | 1                 | Batch                    | Fast   | High      | Low             |

---

**simulate a small example** showing exactly how updates differ in these three cases using a toy dataset. That would make the effect very clear.


```python
import numpy as np
import matplotlib.pyplot as plt

# Simulate simple linear regression y = 2x + 1 with noise
np.random.seed(42)
X = np.random.rand(100, 1)
y = 2 * X + 1 + 0.1 * np.random.randn(100, 1)

# Initialize parameters
def initialize_params():
    w = np.random.randn()
    b = np.random.randn()
    return w, b

# Prediction
def predict(X, w, b):
    return w * X + b

# Compute gradient
def compute_gradients(X_batch, y_batch, w, b):
    m = len(X_batch)
    y_pred = predict(X_batch, w, b)
    dw = (1/m) * np.sum((y_pred - y_batch) * X_batch)
    db = (1/m) * np.sum(y_pred - y_batch)
    return dw, db

# Training function
def train(X, y, batch_size, epochs=5, lr=0.1):
    w, b = initialize_params()
    history = []
    m = len(X)
    for epoch in range(epochs):
        # Shuffle data
        indices = np.arange(m)
        np.random.shuffle(indices)
        X_shuffled = X[indices]
        y_shuffled = y[indices]

        for i in range(0, m, batch_size):
            X_batch = X_shuffled[i:i+batch_size]
            y_batch = y_shuffled[i:i+batch_size]

            dw, db = compute_gradients(X_batch, y_batch, w, b)
            w -= lr * dw
            b -= lr * db

        # Record loss
        y_pred = predict(X, w, b)
        loss = np.mean((y_pred - y) ** 2)
        history.append(loss)
    return history

# Run simulations
batch_sizes = [1, 50, 100]
results = {bs: train(X, y, bs) for bs in batch_sizes}

# Plot
plt.figure(figsize=(8, 5))
for bs, loss_history in results.items():
    plt.plot(loss_history, label=f'Batch Size = {bs}')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.title('Training Loss for Different Batch Sizes')
plt.legend()
plt.grid(True)
plt.show()
```

**Simulation result:**

* **Batch size = 1** → More fluctuation in loss but can adjust weights frequently.
* **Batch size = 50** → Smoother convergence, balanced speed and stability.
* **Batch size = 100** → Very smooth loss curve but fewer weight updates per epoch.

## 6\. Backpropagation

Backpropagation is the algorithm that efficiently calculates the gradients needed for gradient descent. It works by propagating the error backwards through the network, from the output layer to the input layer. Using the **chain rule of calculus**, it determines how much each weight and bias contributed to the overall error, allowing for precise adjustments during optimization.

## 7\. Deep Learning for Regression

### Example Data

Let's generate some synthetic non-linear data for a regression task. Suppose we want to predict a target variable `y` based on an input feature `x`, where `y` has a non-linear relationship with `x` (e.g., a sine wave with some noise).

### Code Example (Keras/TensorFlow)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import mean_squared_error, r2_score

print("--- Deep Learning for Regression ---")

# 1. Generate Synthetic Data
np.random.seed(42)
X_reg = np.linspace(-3, 3, 1000).reshape(-1, 1) # Single feature
y_reg = np.sin(X_reg * 2) + np.random.normal(0, 0.2, X_reg.shape) # Non-linear relationship with noise

# Split data
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# Scale features (important for neural networks)
scaler_X_reg = StandardScaler()
X_train_reg_scaled = scaler_X_reg.fit_transform(X_train_reg)
X_test_reg_scaled = scaler_X_reg.transform(X_test_reg)

# No need to scale y for MSE loss, but can be done if desired and then inverse-transformed

print(f"Training data shape (X): {X_train_reg_scaled.shape}, (y): {y_train_reg.shape}")
print(f"Test data shape (X): {X_test_reg_scaled.shape}, (y): {y_test_reg.shape}")

# 2. Build the Neural Network Model
model_reg = Sequential([
    Dense(64, activation='relu', input_shape=(X_train_reg_scaled.shape[1],)), # Input layer + 1st hidden layer
    Dense(32, activation='relu'), # 2nd hidden layer
    Dense(1) # Output layer: 1 neuron for regression, no activation (linear activation by default)
])

# 3. Compile the Model
# Optimizer: Adam (an advanced version of Gradient Descent)
# Loss: Mean Squared Error (common for regression)
model_reg.compile(optimizer=Adam(learning_rate=0.01), loss='mse')

model_reg.summary()

# 4. Train the Model
history_reg = model_reg.fit(
    X_train_reg_scaled, y_train_reg,
    epochs=100, # Number of passes through the entire dataset
    batch_size=32, # Number of samples per gradient update
    validation_split=0.1, # Use 10% of training data for validation during training
    verbose=0 # Suppress verbose output during training
)

# 5. Evaluate the Model
y_pred_reg_scaled = model_reg.predict(X_test_reg_scaled)

mse = mean_squared_error(y_test_reg, y_pred_reg_scaled)
r2 = r2_score(y_test_reg, y_pred_reg_scaled)

print(f"\nMean Squared Error on Test Set: {mse:.4f}")
print(f"R-squared on Test Set: {r2:.4f}")

# Plotting the results
plt.figure(figsize=(10, 6))
plt.scatter(X_test_reg, y_test_reg, label='Actual Data', alpha=0.6)
# Predict across the full original range for a smooth curve
X_plot_reg = np.linspace(X_reg.min(), X_reg.max(), 500).reshape(-1, 1)
X_plot_reg_scaled = scaler_X_reg.transform(X_plot_reg)
y_plot_pred_reg = model_reg.predict(X_plot_reg_scaled)
plt.plot(X_plot_reg, y_plot_pred_reg, color='red', label='DNN Prediction', linewidth=2)
plt.title('Deep Neural Network for Regression')
plt.xlabel('Feature X')
plt.ylabel('Target Y')
plt.legend()
plt.grid(True)
plt.show()

# Plot training history
plt.figure(figsize=(8, 5))
plt.plot(history_reg.history['loss'], label='Train Loss')
plt.plot(history_reg.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```

## 8\.Deep Learning for Classification

### Example Data

Let's use the famous Iris dataset for a multi-class classification problem. We'll classify iris flowers into one of three species based on their sepal and petal measurements.

### Code Example (Keras/TensorFlow)

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import seaborn as sns

print("\n--- Deep Learning for Classification ---")

# 1. Load and Prepare Data
iris = datasets.load_iris()
X_cls = iris.data # All four features
y_cls = iris.target # 0, 1, 2 for the three species

# Split data
X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls, y_cls, test_size=0.2, random_state=42, stratify=y_cls # Stratify to maintain class proportions
)

# Scale features
scaler_X_cls = StandardScaler()
X_train_cls_scaled = scaler_X_cls.fit_transform(X_train_cls)
X_test_cls_scaled = scaler_X_cls.transform(X_test_cls)

# One-hot encode target variable for multi-class classification
encoder = OneHotEncoder(sparse_output=False) # sparse_output=False for dense array output
y_train_cls_encoded = encoder.fit_transform(y_train_cls.reshape(-1, 1))
y_test_cls_encoded = encoder.transform(y_test_cls.reshape(-1, 1))

num_classes = y_train_cls_encoded.shape[1]

print(f"Training data shape (X): {X_train_cls_scaled.shape}, (y): {y_train_cls_encoded.shape}")
print(f"Test data shape (X): {X_test_cls_scaled.shape}, (y): {y_test_cls_encoded.shape}")
print(f"Number of classes: {num_classes}")

# 2. Build the Neural Network Model
model_cls = Sequential([
    Dense(64, activation='relu', input_shape=(X_train_cls_scaled.shape[1],)), # Input layer + 1st hidden layer
    Dense(32, activation='relu'), # 2nd hidden layer
    Dense(num_classes, activation='softmax') # Output layer: num_classes neurons, softmax for probabilities
])

# 3. Compile the Model
# Loss: Categorical Cross-Entropy (for multi-class classification with one-hot encoded targets)
model_cls.compile(optimizer=Adam(learning_rate=0.01), loss='categorical_crossentropy', metrics=['accuracy'])

model_cls.summary()

# 4. Train the Model
history_cls = model_cls.fit(
    X_train_cls_scaled, y_train_cls_encoded,
    epochs=100,
    batch_size=32,
    validation_split=0.1,
    verbose=0
)

# 5. Evaluate the Model
loss, accuracy = model_cls.evaluate(X_test_cls_scaled, y_test_cls_encoded, verbose=0)
print(f"\nTest Loss: {loss:.4f}")
print(f"Test Accuracy: {accuracy:.4f}")

y_pred_prob_cls = model_cls.predict(X_test_cls_scaled)
y_pred_cls = np.argmax(y_pred_prob_cls, axis=1) # Convert probabilities to class labels

print("\nClassification Report:\n", classification_report(y_test_cls, y_pred_cls, target_names=iris.target_names))

# Confusion Matrix
cm = confusion_matrix(y_test_cls, y_pred_cls)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix for Deep Neural Network Classification')
plt.show()

# Plot training history
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.plot(history_cls.history['loss'], label='Train Loss')
plt.plot(history_cls.history['val_loss'], label='Validation Loss')
plt.title('Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(history_cls.history['accuracy'], label='Train Accuracy')
plt.plot(history_cls.history['val_accuracy'], label='Validation Accuracy')
plt.title('Training and Validation Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 9\.Explanation of Code Examples

1.  **Data Generation/Loading and Preprocessing**:

      * **Regression**: We create `X_reg` (a single feature) and `y_reg` (a target with a `sin` relationship + noise).
      * **Classification**: We load the `iris` dataset, which has 4 features and 3 classes.
      * **Train-Test Split**: Essential to evaluate the model on unseen data.
      * **Scaling**: `StandardScaler` is used to normalize features. This is crucial for neural networks because it helps the optimization algorithm converge faster and prevents features with larger values from dominating the learning process.
      * **One-Hot Encoding (for Classification)**: For multi-class classification, target labels (e.g., 0, 1, 2) are converted into a binary vector format (e.g., [1,0,0], [0,1,0], [0,0,1]). This is required for `categorical_crossentropy` loss.

2.  **Building the Neural Network (`Sequential` model in Keras)**:

      * `Sequential` is a linear stack of layers.
      * `Dense` layers are **fully connected** layers, meaning every neuron in one layer is connected to every neuron in the next layer.
      * `input_shape`: Only specified for the *first* layer to tell the network the shape of your input data.
      * `activation`: Specifies the activation function for the layer.
          * For hidden layers, `relu` is a common choice.
          * For the **output layer in regression**, a linear activation (default for `Dense` if `activation` is not specified) is used, as we want to predict a continuous value.
          * For the **output layer in multi-class classification**, `softmax` is used to get probability distributions over the classes.
          * For **binary classification**, `sigmoid` would be used in the output layer.

3.  **Compiling the Model (`model.compile`)**:

      * `optimizer`: `Adam` is a popular choice for its efficiency and ability to handle different learning rates for different parameters. The `learning_rate` hyperparameter controls the step size of the optimization.
      * `loss`: The loss function that the model will try to minimize. We chose `mse` for regression and `categorical_crossentropy` for multi-class classification.
      * `metrics`: Metrics to monitor during training and evaluation (e.g., `accuracy` for classification).

4.  **Training the Model (`model.fit`)**:

      * `epochs`: The number of times the training algorithm will iterate over the entire training dataset. More epochs can lead to better learning but also overfitting.
      * `batch_size`: The number of samples processed before the model's parameters are updated. Mini-batch gradient descent is used.
      * `validation_split`: A portion of the training data is held out to monitor the model's performance on unseen data during training, helping to detect overfitting early.

5.  **Evaluation (`model.evaluate`, `model.predict`)**:

      * `model.evaluate` provides the final loss and specified metrics on the test set.
      * `model.predict` generates predictions. For classification, these are probabilities, which are then converted to class labels using `np.argmax`.
      * Evaluation metrics like MSE, R-squared for regression, and Accuracy, Classification Report, Confusion Matrix for classification, provide insights into model performance.
