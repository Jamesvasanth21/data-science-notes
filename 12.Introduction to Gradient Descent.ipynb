**Gradient Descent Explained with Visualizations and Small Dataset**

---

### 1. Introduction to Gradient Descent
Gradient Descent is like a step-by-step guide for finding the best line (in linear regression) or curve (in logistic regression) that fits our data. Think of it as climbing down a hill to find the lowest point — the place where our model makes the fewest mistakes.

We use it to improve how our model learns from data by reducing something called "loss" or "error" — which tells us how wrong our model is.

---

### 2. Formula (Beginner Friendly)

Let’s break it down:
\[ \theta := \theta - \alpha \cdot \nabla J(\theta) \]
- \( \theta \) means the weights or parameters the model is learning.
- \( \alpha \) is the learning rate — a small number that decides how big our steps are.
- \( J(\theta) \) is the cost or loss — how bad our predictions are.
- \( \nabla J(\theta) \) is the gradient — the direction we need to go to reduce the error.

We keep adjusting our weights until we reach a point where the cost is very small.

**📊 Visualization Tip:**
Show a simple graph with a U-shaped curve representing the cost function, and a dot rolling downhill representing the steps of gradient descent.

---

### 3. Using a Small Dataset (Linear Regression Example)
Suppose we have a tiny dataset:

| x | y |
|---|---|
| 1 | 1 |
| 2 | 2 |
| 3 | 3 |
| 4 | 4 |

We want to draw the best line \( y = wx + b \). Initially, we start with guesses like \( w = 0, b = 0 \).

**Our Goal:** Make the line as close to the data points as possible.

**Error (Loss) Function — Mean Squared Error (MSE):**
\[ J(w, b) = \frac{1}{n} \sum_{i=1}^{n} (y_i - (w x_i + b))^2 \]

We take small steps by calculating how much to change \( w \) and \( b \):
\[ \frac{\partial J}{\partial w} = -\frac{2}{n} \sum x_i (y_i - \hat{y}_i) \]
\[ \frac{\partial J}{\partial b} = -\frac{2}{n} \sum (y_i - \hat{y}_i) \]

By repeating this process for 10-20 steps, we will see the line improving.

### 6. Logistic Regression: Binary Classification
New dataset:

| x | y |
|---|---|
| 1 | 0 |
| 2 | 0 |
| 3 | 1 |
| 4 | 1 |

We now want to draw a curve that separates 0s from 1s.

Model: \( \hat{y} = \sigma(wx + b) \) where \( \sigma(z) = \frac{1}{1 + e^{-z}} \) is the sigmoid function.

**Loss Function (Cross Entropy):**
\[ J(w, b) = -\frac{1}{n} \sum_{i=1}^n \left[ y_i \log(\hat{y}_i) + (1 - y_i)\log(1 - \hat{y}_i) \right] \]

**Gradient Steps:**
\[ \frac{\partial J}{\partial w} = \frac{1}{n} \sum (\hat{y}_i - y_i)x_i \]
\[ \frac{\partial J}{\partial b} = \frac{1}{n} \sum (\hat{y}_i - y_i) \]

**What to Visualize:**
- Plot a sigmoid curve that goes from 0 to 1.
- Mark 0 and 1 class data points on the x-axis.
- Show how the curve moves as model learns.

**📈 Visualization Tip:**
Use a plot showing the sigmoid curve adjusting after each iteration. Shade the region where the model predicts class 1 vs class 0.

---

### 7. Learning Curve
- Plot cost after each step.
- The curve should go downward — showing learning.
- If the line flattens, learning has stopped (converged).

**📊 Visualization Tip:**
Draw a line graph showing cost on the y-axis and number of iterations on the x-axis. Add markers to show how cost decreases steadily.

---

### 8. Summary (Beginner Focused)
- Gradient Descent helps a model learn by reducing its mistakes step by step.
- Start with a small dataset so you can follow each calculation.
- Visualizations like lines, curves, cost surfaces, and learning graphs make it easier to see what's happening.
- Great for understanding how both linear and logistic regression models improve over time.

## Generate synthetic data

Create a small, simple dataset for linear regression.

```python
import numpy as np

X = np.array([1, 2, 3, 4])
y = np.array([1, 2, 3, 4])

print("X:", X)
print("y:", y)
```

## Initialize parameters

### Subtask:
Start with initial values for the slope (`m`) and y-intercept (`b`).


**Reasoning**:
Initialize the slope and y-intercept variables and print their initial values.



```python
m = 0
b = 0

print("Initial slope (m):", m)
print("Initial y-intercept (b):", b)
```

## Implement gradient descent

### Subtask:
Implement gradient descent to update the parameters `m` and `b` based on the calculated gradients.


**Reasoning**:
Implement the gradient descent function as described in the instructions, including the loop for iterations, gradient calculations, parameter updates, cost calculation, and storing the history of parameters and cost. Then call the function and store the returned values and histories.



```python
def gradient_descent(X, y, learning_rate, n_iterations):
    m = 0
    b = 0
    n = len(X)

    cost_history = []
    m_history = []
    b_history = []

    for i in range(n_iterations):
        y_pred = m * X + b

        # Calculate gradients
        dm = (-2/n) * np.sum(X * (y - y_pred))
        db = (-2/n) * np.sum(y - y_pred)

        # Update parameters
        m = m - learning_rate * dm
        b = b - learning_rate * db

        # Calculate cost (MSE)
        cost = (1/n) * np.sum((y - y_pred)**2)

        # Store history
        cost_history.append(cost)
        m_history.append(m)
        b_history.append(b)

    return m, b, cost_history, m_history, b_history

# Call the gradient descent function
learning_rate = 0.01
n_iterations = 100
final_m, final_b, cost_history, m_history, b_history = gradient_descent(X, y, learning_rate, n_iterations)

print("Final m:", final_m)
print("Final b:", final_b)
print("Cost history length:", len(cost_history))
```

## Visualize the cost surface and parameter updates

### Subtask:
Create a 3D plot to show the cost function for different values of `m` and `b`, and superimpose the path taken by the parameters during gradient descent.


**Reasoning**:
Generate a grid of m and b values, calculate the cost for each combination, and then create and display a 3D surface plot of the cost function with the gradient descent path.



```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create a grid of m and b values
m_values = np.linspace(-1, 3, 100)
b_values = np.linspace(-1, 3, 100)
M, B = np.meshgrid(m_values, b_values)

# Calculate the cost for each combination of m and b
cost_grid = np.zeros(M.shape)
n = len(X)
for i in range(M.shape[0]):
    for j in range(M.shape[1]):
        y_pred_grid = M[i, j] * X + B[i, j]
        cost_grid[i, j] = (1/n) * np.sum((y - y_pred_grid)**2)

# Create the 3D plot
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
ax.plot_surface(M, B, cost_grid, cmap='viridis', alpha=0.8)

# Plot the gradient descent path
ax.plot(m_history, b_history, cost_history, marker='o', color='r', markersize=5, label='Gradient Descent Path')

# Add labels and title
ax.set_xlabel("Slope (m)")
ax.set_ylabel("Y-intercept (b)")
ax.set_zlabel("Cost (MSE)")
ax.set_title("Cost Surface and Gradient Descent Path")
ax.legend()

# Display the plot
plt.show()
```

# Derive Gradients:

Derive **gradients** of a linear function $y = mx + b$ with respect to $m$ and $b$, in the context of **loss minimization**

---

### Step 1: The function

$$
y = mx + b
$$

* $m$ = slope (weight)
* $b$ = intercept (bias)
* $x$ = input feature
* $y$ = prediction

---

### Step 2: Loss function

Suppose we’re doing supervised learning with true value $y_{\text{true}}$.
We define a **loss function** $L$. A common choice is the **Mean Squared Error (MSE):**

$$
L = \frac{1}{2}(y_{\text{pred}} - y_{\text{true}})^2
$$

We add the $\tfrac{1}{2}$ just to make derivatives cleaner.

Here:

$$
y_{\text{pred}} = mx + b
$$

So:

$$
L = \frac{1}{2} \big( (mx+b) - y_{\text{true}} \big)^2
$$

---

### Step 3: Derivative wrt $m$ ($\frac{\partial L}{\partial m}$)

Using chain rule:

$$
\frac{\partial L}{\partial m} = \frac{\partial L}{\partial y_{\text{pred}}} \cdot \frac{\partial y_{\text{pred}}}{\partial m}
$$

1.

$$
\frac{\partial L}{\partial y_{\text{pred}}} = (y_{\text{pred}} - y_{\text{true}})
$$

2.

$$
\frac{\partial y_{\text{pred}}}{\partial m} = \frac{\partial (mx+b)}{\partial m} = x
$$

So:

$$
\frac{\partial L}{\partial m} = (y_{\text{pred}} - y_{\text{true}}) \cdot x
$$

---

### Step 4: Derivative wrt $b$ ($\frac{\partial L}{\partial b}$)

Again, chain rule:

$$
\frac{\partial L}{\partial b} = \frac{\partial L}{\partial y_{\text{pred}}} \cdot \frac{\partial y_{\text{pred}}}{\partial b}
$$

1.

$$
\frac{\partial L}{\partial y_{\text{pred}}} = (y_{\text{pred}} - y_{\text{true}})
$$

2.

$$
\frac{\partial y_{\text{pred}}}{\partial b} = 1
$$

So:

$$
\frac{\partial L}{\partial b} = (y_{\text{pred}} - y_{\text{true}})
$$

---

### Step 5: Interpretation

* $\frac{\partial L}{\partial m} = (y_{\text{pred}} - y_{\text{true}})\cdot x$
  👉 The error scaled by the input $x$. If $x$ is large, mistakes affect the slope more strongly.

* $\frac{\partial L}{\partial b} = (y_{\text{pred}} - y_{\text{true}})$
  👉 The error itself directly adjusts the intercept.

---

### Step 6: Update rule (Gradient Descent)

If learning rate = $\alpha$:

$$
m_{\text{new}} = m - \alpha \cdot \frac{\partial L}{\partial m}
$$

$$
b_{\text{new}} = b - \alpha \cdot \frac{\partial L}{\partial b}
$$

---

🔑 Intuition:

* $dl/dm$ → "How much the slope $m$ should move to reduce the error."
* $dl/db$ → "How much the intercept $b$ should move to reduce the error."
