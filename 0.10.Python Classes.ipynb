# Python Classes and Object-Oriented Programming (OOP)

Welcome to an exciting part of your Python journey: **Classes and Object-Oriented Programming (OOP)\!** So far, you've learned about variables, data types, loops, functions, and various data structures. These are like learning the individual words and sentences of a language.

Now, we're going to learn how to organize those words and sentences into coherent paragraphs, chapters, and even entire books\! OOP is a powerful way to structure your code, making it more organized, reusable, and easier to understand and maintain.

### What is Object-Oriented Programming (OOP)?

Imagine you're building a video game or a system for a school.

  * Instead of just having separate variables for `player_health`, `player_score`, `enemy_health`, `enemy_position`, etc., OOP allows you to create "blueprints" for things like `Player` or `Enemy`.
  * These blueprints define what attributes (data) an object has and what actions (functions) it can perform.

In OOP, we think about programs as collections of **objects** that interact with each other.

  * **Object:** A real-world entity, like a specific `Car`, a `Student`, or a `Book`.
  * **Class:** A blueprint or a template for creating objects. It defines the common characteristics (attributes) and behaviors (methods/functions) that all objects of that type will have.

Let's dive into how this works in Python\!

## 1\. What are Classes and Objects?

### Analogy: Cookies and Cookie Cutters

Think about making cookies:

  * The **cookie cutter** is like a **Class**. It's not a cookie itself, but it defines the *shape* and *structure* of any cookie you make with it.
  * The actual **cookies** you bake are like **Objects**. Each cookie is a distinct item, but it was created using the same cookie cutter. Each cookie has its own unique properties (e.g., this one has a bite taken out, that one has sprinkles) even though they share the same basic shape.


### Class: The Blueprint

In Python, you define a class using the `class` keyword.

```
# This is our blueprint for a Car
class Car:
    # Attributes (characteristics) that all Cars will have
    # We'll define these specifically when we create a Car object

    # Methods (actions/behaviors) that a Car can perform
    def start_engine(self):
        print("Engine started!")

    def drive(self):
        print("Car is driving.")

    def honk(self):
        print("Honk! Honk!")

print("Car class defined.")
```

### Object: An Instance of the Blueprint

An object is an **instance** of a class. You create an object by calling the class name as if it were a function.

```
# Creating objects (instances) from the Car class
my_car = Car()  # my_car is now an object of the Car class
your_car = Car() # your_car is another object of the Car class

print(f"Type of my_car: {type(my_car)}")
print(f"Is my_car an instance of Car? {isinstance(my_car, Car)}")
```

Now that we have objects, we can make them *do* things defined by their class's methods\!

```
print("\n--- my_car actions ---")
my_car.start_engine()
my_car.drive()
my_car.honk()

print("\n--- your_car actions ---")
your_car.drive() # Both cars can drive
your_car.honk()
```

Notice that `my_car` and `your_car` are separate objects. Even though they came from the same `Car` blueprint, they are distinct entities.

## 2\. Attributes (Data) and the `__init__` Method

Objects usually have **attributes** (data or properties) that describe them. For a `Car`, these might be `make`, `model`, `year`, `color`.

When you create an object, you often want to give it initial attributes. This is done using a special method called `__init__` (pronounced "dunder init" - "dunder" for double underscore).

### The `__init__` Method (Constructor)

  * The `__init__` method is automatically called **every time a new object (instance) of the class is created**.
  * Its purpose is to **initialize the attributes** of the new object.
  * The first parameter of any method in a class, including `__init__`, is always `self`.

#### What is `self`?

`self` is a convention (not a keyword, but universally used) that refers to the **instance of the class itself**. When you call a method like `my_car.drive()`, Python automatically passes `my_car` as the `self` argument to the `drive` method.

```
class Car:
    # The __init__ method for initializing attributes
    def __init__(self, make, model, year, color):
        # 'self.make' refers to the 'make' attribute of this specific Car object
        self.make = make
        self.model = model
        self.year = year
        self.color = color
        self.is_engine_on = False # All new cars start with engine off

    def start_engine(self):
        if not self.is_engine_on:
            print(f"The {self.color} {self.make} {self.model}'s engine started!")
            self.is_engine_on = True
        else:
            print("Engine is already on.")

    def drive(self):
        if self.is_engine_on:
            print(f"The {self.make} {self.model} is driving.")
        else:
            print("Cannot drive. Engine is off. Start the engine first!")

    def honk(self):
        print("Honk! Honk!")

# Now, when we create a Car object, we must provide these initial values
my_new_car = Car("Toyota", "Camry", 2023, "Silver")
friends_car = Car("Honda", "Civic", 2022, "Blue")

print(f"\nMy new car details:")
print(f"Make: {my_new_car.make}")
print(f"Model: {my_new_car.model}")
print(f"Year: {my_new_car.year}")
print(f"Color: {my_new_car.color}")
print(f"Engine status: {my_new_car.is_engine_on}") # Initially False

print(f"\nFriend's car details:")
print(f"Make: {friends_car.make}")
print(f"Model: {friends_car.model}")
print(f"Year: {friends_car.year}")
print(f"Color: {friends_car.color}")

my_new_car.drive() # Try to drive before starting
my_new_car.start_engine()
my_new_car.drive()
my_new_car.start_engine() # Try to start again
```

## 3\. Methods (Functions within a Class)

**Methods** are functions defined inside a class. They describe the behaviors or actions that objects of that class can perform.

  * Methods always have `self` as their first parameter.
  * Methods can access and modify the object's attributes using `self.attribute_name`.

Let's refine our `Car` class with more methods that interact with its attributes.

```
class Car:
    def __init__(self, make, model, year, color, fuel_level_liters):
        self.make = make
        self.model = model
        self.year = year
        self.color = color
        self.fuel_level_liters = fuel_level_liters
        self.is_engine_on = False

    def start_engine(self):
        if self.fuel_level_liters <= 0:
            print("Cannot start engine, fuel tank is empty!")
            return False
        if not self.is_engine_on:
            print(f"The {self.color} {self.make} {self.model}'s engine started!")
            self.is_engine_on = True
            return True
        else:
            print("Engine is already on.")
            return False

    def drive(self, distance_km):
        if not self.is_engine_on:
            print("Cannot drive. Engine is off. Start the engine first!")
            return

        fuel_needed = distance_km * 0.1 # Assume 0.1 liters/km
        if self.fuel_level_liters >= fuel_needed:
            self.fuel_level_liters -= fuel_needed
            print(f"Drove {distance_km} km. Remaining fuel: {self.fuel_level_liters:.2f} liters.")
        else:
            print(f"Not enough fuel to drive {distance_km} km. You only have {self.fuel_level_liters:.2f} liters.")

    def add_fuel(self, liters):
        if liters > 0:
            self.fuel_level_liters += liters
            print(f"Added {liters} liters of fuel. New fuel level: {self.fuel_level_liters:.2f} liters.")
        else:
            print("Please add a positive amount of fuel.")

    def get_info(self):
        """Returns a string with car information."""
        return (f"Car: {self.year} {self.make} {self.model}, Color: {self.color}, "
                f"Fuel: {self.fuel_level_liters:.2f}L, Engine On: {self.is_engine_on}")

# Create a car with some fuel
my_sports_car = Car("Ferrari", "488 GTB", 2024, "Red", 50)
print(my_sports_car.get_info())

my_sports_car.start_engine()
my_sports_car.drive(200) # Drive 200 km
print(my_sports_car.get_info())

my_sports_car.drive(400) # Try to drive more
print(my_sports_car.get_info())

my_sports_car.add_fuel(30)
print(my_sports_car.get_info())
my_sports_car.drive(400) # Now it should be able to drive
print(my_sports_car.get_info())
```

## 4\. `self` in Detail: Why it's Important

Let's re-emphasize `self`. It's the mechanism that connects the methods to the specific object they are operating on. Without `self`, a method wouldn't know which object's `make`, `model`, or `fuel_level_liters` it should be using or changing.

```
class Dog:
    def __init__(self, name, breed):
        self.name = name # 'self.name' belongs to this specific Dog object
        self.breed = breed

    def bark(self):
        # 'self.name' accesses the name of the specific Dog object calling this method
        print(f"{self.name} says Woof!")

    def describe(self):
        print(f"{self.name} is a {self.breed}.")

dog1 = Dog("Buddy", "Golden Retriever")
dog2 = Dog("Lucy", "Beagle")

dog1.bark()    # When dog1.bark() is called, 'self' inside bark() refers to dog1
dog2.bark()    # When dog2.bark() is called, 'self' inside bark() refers to dog2

dog1.describe()
dog2.describe()
```

## 5\. Benefits of OOP

Why bother with classes and objects?

1.  **Organization:** Code is structured logically into classes, making it easier to find and understand related pieces of code.
2.  **Modularity:** Each class is a self-contained unit. You can work on one class without affecting others, promoting easier development by multiple team members.
3.  **Reusability:** Once you define a `Car` class, you can create as many `Car` objects as you need, without rewriting the common code for each.
4.  **Maintainability:** Changes or bug fixes to a class only need to be made in one place (the class definition), and all objects created from it will inherit those changes.
5.  **Abstraction:** You can focus on what an object *does* rather than how it *does* it. (e.g., `my_car.drive()` abstracts away the complex mechanics of driving).

## 6\. Real-World OOP Example: A Simple Bank Account

Let's put it all together to create a simple `BankAccount` class.

```
class BankAccount:
    def __init__(self, account_holder, initial_balance=0):
        # Initialize attributes for the bank account
        self.account_holder = account_holder
        # Ensure initial balance is not negative
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        self.balance = initial_balance
        print(f"Account created for {self.account_holder} with initial balance {self.balance:.2f}.")

    def deposit(self, amount):
        """Deposits a positive amount to the account."""
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount:.2f}. New balance: {self.balance:.2f}.")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        """Withdraws a positive amount from the account if funds are sufficient."""
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self.balance:
            print(f"Insufficient funds. Available: {self.balance:.2f}, Requested: {amount:.2f}.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount:.2f}. New balance: {self.balance:.2f}.")

    def get_balance(self):
        """Returns the current balance."""
        return self.balance

    def get_account_info(self):
        """Prints all account details."""
        print(f"\n--- Account Information ---")
        print(f"Account Holder: {self.account_holder}")
        print(f"Current Balance: {self.balance:.2f}")
        print(f"--------------------------")

# Create some bank accounts
johns_account = BankAccount("John Doe", 500.00)
janes_account = BankAccount("Jane Smith") # Default initial balance is 0

johns_account.deposit(100.50)
johns_account.withdraw(200.00)
johns_account.withdraw(500.00) # Attempt to overdraw

janes_account.deposit(75.25)
janes_account.withdraw(10.00)

johns_account.get_account_info()
janes_account.get_account_info()

print(f"John's final balance: {johns_account.get_balance():.2f}")
```
