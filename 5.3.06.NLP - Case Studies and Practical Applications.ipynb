## 1. Real-world NLP Projects

Natural Language Processing (NLP) is widely used in various industries such as healthcare, finance, customer support, and social media.

### ✅ Example 1: Sentiment Analysis on Twitter Data

**Use Case:**  
Analyze public sentiment about a product, event, or service.

**Libraries:** `tweepy`, `nltk`, `TextBlob`


```python
import tweepy
from textblob import TextBlob

# Twitter API credentials (replace with your own)
consumer_key = 'YOUR_KEY'
consumer_secret = 'YOUR_SECRET'
access_token = 'YOUR_ACCESS_TOKEN'
access_token_secret = 'YOUR_ACCESS_SECRET'

# Authenticate to Twitter
auth = tweepy.OAuth1UserHandler(consumer_key, consumer_secret, access_token, access_token_secret)
api = tweepy.API(auth)

# Get tweets about 'AI'
tweets = api.search_tweets(q="AI", count=100)

# Analyze sentiment
for tweet in tweets:
    analysis = TextBlob(tweet.text)
    print(f"Tweet: {tweet.text}")
    print(f"Sentiment: {analysis.sentiment}\n")
```

**Useful Links:**

* [TextBlob Documentation](https://textblob.readthedocs.io/en/dev/)
* [Tweepy Documentation](https://docs.tweepy.org/en/stable/)

### ✅ Example 2: Named Entity Recognition (NER)

**Use Case:**
Extract entities like person names, locations, and organizations from text.

**Libraries:** `spacy`

```python
import spacy

# Load the pre-trained model
nlp = spacy.load("en_core_web_sm")

text = "Apple is opening a new office in Bangalore with CEO Tim Cook attending the launch."

doc = nlp(text)

for ent in doc.ents:
    print(ent.text, ent.label_)
```


**Useful Links:**

* [spaCy Documentation](https://spacy.io/usage/spacy-101)

## 2. Challenges and Solutions in NLP Applications

### 🚧 Challenge 1: Ambiguity in Language

**Problem:**
Words like "bank" could refer to a financial institution or the side of a river.

**Solution:**
Use contextual embeddings like BERT.

```python
from transformers import pipeline

classifier = pipeline("zero-shot-classification")

sequence = "The bank will loan money to new customers."

labels = ["finance", "geography", "sports"]

result = classifier(sequence, labels)

print(result)
```
