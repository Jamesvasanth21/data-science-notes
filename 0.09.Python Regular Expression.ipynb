# Python Regular Expressions: A Powerful Tool for Text Processing

## Introduction

Welcome to Regular Expressions (often shortened to "RegEx" or "Regex") in Python! If you've ever needed to search for specific patterns in text, validate input formats, or replace certain strings, regular expressions are your best friend. They provide a concise and flexible way to find and manipulate strings of text based on patterns.

While they might look a bit intimidating at first glance, learning the basics of RegEx will significantly boost your text processing capabilities in Python.

---

## 1. What are Regular Expressions?

A **Regular Expression** is a sequence of characters that defines a search pattern. When you search for a pattern in text, you can use these patterns to describe what you are looking for.

Think of it as a super-powered version of the "Find" functionality you might use in a word processor. Instead of just finding exact words, you can find patterns like:

* "Any phone number"
* "All email addresses"
* "Dates in the format DD-MM-YYYY"
* "Words that start with 'a' and end with 'e'"

In Python, the `re` module provides full support for regular expressions.

### Basic Concepts:

* **Characters**: Most characters match themselves (e.g., `a` matches `a`, `1` matches `1`).
* **Metacharacters**: Special characters that have a predefined meaning and don't match themselves (e.g., `.`, `*`, `+`, `?`, `[]`, `\`, `^`, `$`). We'll explore these soon!
* **Raw Strings (`r""`)**: It's highly recommended to use raw strings for regular expressions in Python. A raw string treats backslashes (`\`) as literal characters, which prevents Python from interpreting them as escape sequences. This is crucial because backslashes are heavily used in RegEx for special sequences.

````
# Normal string: \n is a newline character
print("Hello\nWorld")

# Raw string: \n is treated as literal \ and n
print(r"Hello\nWorld")

# This is important for regex patterns like \d (digit)
# Without r, \d might be interpreted differently
import re
# This might give a SyntaxWarning or errors in some cases
# pattern_bad = "\d+"
# Better:
pattern_good = r"\d+"
print(f"Good pattern: {pattern_good}")
print(f"Pattern:'\d+'")
````

## 2. The `match()` Function and the `search()` Function

The `re` module provides several functions to work with regular expressions. Two of the most commonly used for finding patterns are `re.match()` and `re.search()`.

Both functions return a `Match` object if a match is found, and `None` if no match is found. The `Match` object contains information about the match, such as the matched string and its position.

### `re.match(pattern, string)`

* **Purpose**: Checks for a match **only at the beginning of the string**.
* **Behavior**: If the pattern is found at the very first character of the string, it returns a `Match` object. Otherwise, it returns `None`.

````
import re

text = "Hello World"
pattern = r"Hello"

# Using re.match()
match_object = re.match(pattern, text)

if match_object:
    print(f"Match found at the beginning: {match_object.group()}")
    print(f"Start index: {match_object.start()}")
    print(f"End index: {match_object.end()}")
else:
    print("No match found at the beginning.")

# Example where match() fails because pattern is not at the start
text2 = "World Hello"
match_object2 = re.match(pattern, text2)

if match_object2:
    print(f"Match found: {match_object2.group()}")
else:
    print(f"No match found at the beginning of '{text2}'.") # This will be printed
````

### `re.search(pattern, string)`

* **Purpose**: Scans through the entire string to find the **first occurrence** of the pattern.
* **Behavior**: If the pattern is found anywhere in the string, it returns a `Match` object for the first match. Otherwise, it returns `None`.

````
import re

text = "Hello World"
pattern = r"World"

# Using re.search()
search_object = re.search(pattern, text)

if search_object:
    print(f"Search found anywhere: {search_object.group()}")
    print(f"Start index: {search_object.start()}")
    print(f"End index: {search_object.end()}")
else:
    print("No match found.")

# Example where search() succeeds even if not at the start
text2 = "Learning Python Regular Expressions"
pattern2 = r"Python"
search_object2 = re.search(pattern2, text2)

if search_object2:
    print(f"Search found anywhere in '{text2}': {search_object2.group()}")
    print(f"Start index: {search_object2.start()}")
    print(f"End index: {search_object2.end()}")
else:
    print("No match found.")
````

### Methods of a `Match` Object:

When `match()` or `search()` find a pattern, they return a `Match` object. Here are some useful methods of this object:

* `.group()`: Returns the string matched by the pattern.
* `.start()`: Returns the starting index of the match.
* `.end()`: Returns the ending index (exclusive) of the match.
* `.span()`: Returns a tuple containing the `(start, end)` indices of the match.

````
import re

text = "The quick brown fox jumps over the lazy dog."
pattern = r"fox"

match = re.search(pattern, text)

if match:
    print(f"Matched string: {match.group()}")
    print(f"Start index: {match.start()}")
    print(f"End index: {match.end()}")
    print(f"Span: {match.span()}")
else:
    print("No match found.")
````

## 3. Matching vs. Searching

The key difference between `re.match()` and `re.search()` lies in *where* they look for the pattern:

* **`re.match()`**: Strict about matching from the *beginning* of the string.
* **`re.search()`**: Scans the *entire string* for the first occurrence of the pattern.

Consider this example:

````
import re

my_string = "My favorite fruit is apple."
pattern = r"fruit"

print(f"String: '{my_string}'")
print(f"Pattern: '{pattern}'")

print("\n--- Using re.match() ---")
match_result = re.match(pattern, my_string)
if match_result:
    print(f"Match found by match(): {match_result.group()}")
    print(f"Start index: {match_result.start()}")
    print(f"End index: {match_result.end()}")
else:
    print("re.match() did not find a match because 'fruit' is not at the beginning.")
````

````
print("\n--- Using re.match() first word---")
my_string_ = "fruit is apple."
pattern_ = r"fruit"

match_result_ = re.match(pattern_, my_string_)
if match_result_:
    print(f"Match found by match(): {match_result_.group()}")
    print(f"Start index: {match_result_.start()}")
    print(f"End index: {match_result_.end()}")
else:
    print("re.match() did not find a match because 'fruit' is not at the beginning.")
````

````
print("\n--- Using re.search() ---")
my_string = "My favorite fruit is apple."
pattern = r"fruit"

print(f"String: '{my_string}'")
print(f"Pattern: '{pattern}'")

search_result = re.search(pattern, my_string)
if search_result:
    print(f"Match found by search(): {search_result.group()}")
    print(f"Start index: {search_result.start()}")
    print(f"End index: {search_result.end()}")
else:
    print("re.search() did not find a match.")
````

````
print("\n--- Using re.search() multiple words---")
my_string_ = "apple is a fruit. banana is a fruit. strawberry is a fruit"
pattern_ = r"fruit"

print(f"String: '{my_string_}'")
print(f"Pattern: '{pattern_}'")

match_result_ = re.search(pattern_, my_string_)
if match_result_:
    print(f"Match found by search(): {match_result_.group()}")
    print(f"Start index: {match_result_.start()}")
    print(f"End index: {match_result_.end()}")
else:
    print("re.search() did not find a match because 'fruit' present in string.")
````

**When to use which?**

* Use `re.match()` when you are certain that the pattern should appear at the very beginning of the string.
* Use `re.search()` when you want to find the first occurrence of a pattern anywhere within the string.

## 4. Search and Replace

Regular expressions are not just for finding; they are also incredibly powerful for replacing parts of a string. The `re.sub()` function is used for this.

### `re.sub(pattern, repl, string, count=0, flags=0)`

* `pattern`: The regular expression pattern to search for.
* `repl`: The replacement string.
* `string`: The original string to perform the replacement on.
* `count`: (Optional) The maximum number of pattern occurrences to replace. Default is 0 (replace all occurrences).
* `flags`: (Optional) Modifiers for the pattern (e.g., `re.IGNORECASE`).

````
import re

text = "The quick brown fox jumps over the lazy dog."

# Example 1: Simple replacement
new_text = re.sub(r"fox", "cat", text)
print(f"Original: {text}")
print(f"Replaced 'fox' with 'cat': {new_text}")

# Example 2: Replacing multiple occurrences
text_colors = "Red, Green, Blue, Red, Yellow"
new_colors = re.sub(r"Red", "Pink", text_colors)
print(f"Original colors: {text_colors}")
print(f"Replaced all 'Red' with 'Pink': {new_colors}")

# Example 3: Replacing only the first occurrence
new_colors_first = re.sub(r"Red", "Pink", text_colors, 1)
print(f"Replaced first 'Red' with 'Pink': {new_colors_first}")

# Example 4: Replacing digits with a placeholder
phone_number = "My phone is 123-456-7890."
censored_phone = re.sub(r"\d", "*", phone_number) # \d matches any digit
print(f"Original phone: {phone_number}")
print(f"Censored phone: {censored_phone}")

# Example 5: Replacing with a function
# The repl argument can be a function that takes a match object
def replace_with_double(match):
    print(match.group())
    return str(int(match.group()) * 2)

numbers_text = "The numbers are 5 and 10."
doubled_numbers = re.sub(r"\d+", replace_with_double, numbers_text) # \d+ matches one or more digits
print(f"Original numbers text: {numbers_text}")
print(f"Doubled numbers text: {doubled_numbers}")
````

## 5. Extended Regular Expressions and Wildcards (Metacharacters)

This is where regular expressions become truly powerful! Metacharacters allow you to define complex patterns, not just literal strings.

### Common Metacharacters and Special Sequences:

* `.` (Dot): Matches any single character (except newline).

````
print(re.search(r"c.t", "cat").group()) # cat
print(re.search(r"c.t", "cot").group()) # cot
print(re.search(r"c.t", "cut").group()) # cut
print(re.search(r"c.t", "c8t").group()) # cut
print(re.search(r"c.t", "caat")) # None (too many characters)
````

* `^`: Matches the beginning of the string.

````
print(re.search(r"^Hello", "Hello World").group()) # Hello
print(re.search(r"^World", "Hello World")) # None
````

* `$`: Matches the end of the string.

````
import re
print(re.search(r"World$", "Hello World").group()) # World
print(re.search(r"Worl$", "Hello World")) # None
````

* `*`: Matches zero or more occurrences of the preceding character/group.

````
print(re.search(r"ab*c", "ac").group())    # ac (0 'b's)
print(re.search(r"ab*c", "abc").group())   # abc (1 'b')
print(re.search(r"ab*c", "abbbbc").group()) # abbbbc (4 'b's)
````

* `+`: Matches one or more occurrences of the preceding character/group.

````
print(re.search(r"ab+c", "abc").group())   # abc (1 'b')
print(re.search(r"ab+c", "abbbbc").group()) # abbbbc (4 'b's)
print(re.search(r"ab+c", "ac")) # None (0 'b's)
````

* `?`: Matches zero or one occurrence of the preceding character/group (makes it optional).

````
print(re.search(r"colou?r", "color").group())  # color
print(re.search(r"colou?r", "colour").group()) # colour
print(re.search(r"colou?r", "coloor")) # None
print(re.search(r"colou?r", "colouur")) # None
````

* `{n}`: Matches exactly `n` occurrences of the preceding character/group.

````
print(re.search(r"\d{3}", "123").group())     # 123 (\d matches any digit)
print(re.search(r"\d{3}", "hello123world").group()) # 123
print(re.search(r"\d{3}", "12")) # None
````

* `{n,}`: Matches `n` or more occurrences.

````
print(re.search(r"a{2,}", "aa").group()) # aa
print(re.search(r"a{2,}", "aaaaa").group()) # aaaaa
print(re.search(r"a{2,}", "a")) # None
````

* `{n,m}`: Matches between `n` and `m` (inclusive) occurrences.

````
print(re.search(r"a{2,4}", "aa").group())   # aa
print(re.search(r"a{2,4}", "aaa").group())  # aaa
print(re.search(r"a{2,4}", "aaaa").group()) # aaaa
print(re.search(r"a{2,4}b", "aaaaa")) # aaaa (matches the first 4)
print(re.search(r"a{2,4}", "a")) # None
````

* `[]` (Character Set): Matches any one of the characters inside the brackets.

````
print(re.search(r"[aeiou]", "hello").group()) # e
print(re.search(r"[0-9]", "abc123xyz").group()) # 1 (matches any digit)
print(re.search(r"[A-Z]", "abcDEF").group()) # D (matches any uppercase letter)
print(re.search(r"[aeiou]", "rhythm")) # None
````

* `[^...]`: Matches any character *not* inside the brackets.

````
print(re.search(r"[^aeiou]", "apple").group()) # p (first non-vowel)
print(re.search(r"[^0-9]", "123abc").group()) # a (first non-digit)
````

* `|` (OR): Acts as an OR operator.

````
print(re.search(r"cat|dog", "I have a cat.").group()) # cat
print(re.search(r"cat|dog", "I have a dog.").group()) # dog
print(re.search(r"cat|dog", "I have a bird.")) # None

print(re.search(r"cat|dog", "I have a dog and cat.").group()) # cat
````

````
print(re.search(r"cat&dog", "I have a cat.")) # cat
print(re.search(r"cat&dog", "I have a dog.")) # dog
print(re.search(r"cat&dog", "I have a bird.")) # None

print(re.search(r"cat&dog", "I have a dog&cat.")) # cat
print(re.search(r"cat&dog", "I have a cat&dog.")) # cat
````

* `()` (Grouping): Groups parts of the pattern together. Can be used for "capturing" matched sub-strings or applying quantifiers to a group.

````
# Quantifier on a group
print(re.search(r"(ab)+", "ababab").group()) # ababab

# Capturing groups
match = re.search(r"(\d{4})-(\d{2})-(\d{2})", "Date: 2023-10-26")
if match:
    print(f"Full match: {match.group(0)}") # or match.group()
    print(f"Year: {match.group(1)}")
    print(f"Month: {match.group(2)}")
    print(f"Day: {match.group(3)}")
````

### Special Sequences (Prefixed with `\`)

* `\d`: Matches any digit (0-9). Equivalent to `[0-9]`.
* `\D`: Matches any non-digit. Equivalent to `[^0-9]`.
* `\w`: Matches any alphanumeric character (letters, numbers, underscore). Equivalent to `[a-zA-Z0-9_]`.
* `\W`: Matches any non-alphanumeric character. Equivalent to `[^a-zA-Z0-9_]`.
* `\s`: Matches any whitespace character (space, tab, newline, etc.).
* `\S`: Matches any non-whitespace character.
* `\b`: Matches a word boundary (the position between a word character and a non-word character).
* `\B`: Matches a non-word boundary.

````
import re
# \d (digit)
print(re.findall(r"\d+", "I have 3 apples and 10 bananas.")) # ['3', '10']

# \s (whitespace)
print(re.split(r"\s+", "This    is  a   sentence.")) # ['This', 'is', 'a', 'sentence.']

# \w (word character)
print(re.findall(r"\w+", "Hello, world! How are you? HELLO_WORLD!!")) # ['Hello', 'world', 'How', 'are', 'you']

# \b (word boundary)
text = "The cat sat on the mat."
print(re.findall(r"\bcat\b", text)) # ['cat'] (matches 'cat' as a whole word)
print(re.findall(r"cat", "caterpillar")) # ['cat'] (matches 'cat' within 'caterpillar')
print(re.findall(r"\bcat\b", "caterpillar")) # [] (does not match 'cat' within 'caterpillar')
print(re.findall(r"\bcat", "caterpillar")) # [] (does not match 'cat' within 'caterpillar')
````

### Flags (Modifying RegEx Behavior)

Flags change how the regular expression is interpreted. You pass them as the `flags` argument in `re` functions.

* `re.IGNORECASE` or `re.I`: Performs case-insensitive matching.
* `re.MULTILINE` or `re.M`: Makes `^` and `$` match the start/end of *each line*, not just the entire string.
* `re.DOTALL` or `re.S`: Makes the `.` (dot) metacharacter match *all* characters, including newline characters.

````
import re

# re.IGNORECASE (re.I)
text = "Hello World, hello python"
print(re.findall(r"hello", text))              # ['hello']
print(re.findall(r"hello", text, re.IGNORECASE)) # ['Hello', 'hello']

# ---
# re.MULTILINE (re.M)
text_multi = "Line 1\nLine 2\nLine 3"
print(f"\nOriginal text:\n{text_multi}")
print(f"Lines starting with 'Line' (no multiline): {re.findall(r'^Line', text_multi)}")
print(f"Lines starting with 'Line' (multiline): {re.findall(r'^Line', text_multi, re.MULTILINE)}")

# ---
# re.DOTALL (re.S)
text_with_newline = "First line.\nSecond line."
print(f"\nText with newline:\n{text_with_newline}")

# Without DOTALL, . does not match newline
print(f"Matching 'line.' with just dot: {re.search(r'line.', text_with_newline).group()}")

# CORRECTED LINE: Define the pattern outside the f-string expression
pattern_dotall = r'line.\nSecond line.'
print(f"Matching '{pattern_dotall}' with dotall: {re.search(pattern_dotall, text_with_newline, re.DOTALL).group()}")


# Without DOTALL, . does not match newline
print(f"Matching 'line.' with just dot: {re.search(pattern_dotall, text_with_newline).group()}")
````

## 6. Advanced `re` functions: `findall()` and `split()`

While `match()` and `search()` are for single occurrences, `findall()` and `split()` are for multiple matches or splitting strings.

### `re.findall(pattern, string, flags=0)`

* **Purpose**: Returns a list containing *all non-overlapping matches* of the pattern in the string.

````
import re

text = "Emails: test$@example.com, user@domain.org, another@mail.co.uk"
email_pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"

found_emails = re.findall(email_pattern, text)
print(f"All emails found: {found_emails}")

numbers_text = "The numbers are 12, 45, and 89."
numbers = re.findall(r"\d+", numbers_text)
print(f"All numbers found: {numbers}")
````

### `re.split(pattern, string, maxsplit=0, flags=0)`

* **Purpose**: Splits the string by occurrences of the pattern. Returns a list of strings between the matches.

````
import re

sentence = "This is a sentence with     multiple   spaces."

# Split by one or more spaces
words = re.split(r"\s+", sentence)
print(f"Split by spaces: {words}")

date_info = "Date: 2023-10-26; Time: 15:30:00"
# Split by non-alphanumeric characters (except space)
parts = re.split(r"[^\w\s]+", date_info) # Split by one or more characters that are not word chars or spaces
print(f"Split by non-alphanumeric/non-space: {parts}")

# Limiting splits
data = "apple,banana,cherry,grape"
limited_split = re.split(r",", data, maxsplit=2)
print(f"Limited split (max 2): {limited_split}")
````

### 🔍 Why Use `re.DOTALL`?

The `re.DOTALL` flag in Python’s `re` module is used to make the dot (`.`) match **any character**, including **newline characters** (`\n`). By default, the dot matches any character **except** a newline.

---

By default:

```python
import re

text = "Hello\nWorld"
pattern = re.compile("Hello.World")
print(pattern.search(text))  # Output: None
```

Here, the `.` does **not** match the newline `\n`, so it fails to find `"Hello\nWorld"`.

But with `re.DOTALL`:

```python
pattern = re.compile("Hello.World", re.DOTALL)
print(pattern.search(text))  # Output: <re.Match object; span=(0, 11), match='Hello\nWorld'>
```

Now it works — because `.` can match newline `\n`.

---

### ✅ Syntax

```python
re.compile(pattern, re.DOTALL)
# or use inline flag
re.compile("(?s)pattern")
```

---

### 📘 Examples

#### 1. Matching Multiline Strings

```python
text = """Start
This is the middle.
End"""

pattern = re.compile("Start.*End")        # Default behavior
print(pattern.search(text))               # None

pattern = re.compile("Start.*End", re.DOTALL)
print(pattern.search(text))               # Matches entire string from Start to End
```

Without `re.DOTALL`, `.*` stops at the end of the first line. With `re.DOTALL`, it spans across lines.

---

#### 2. Inline Flag `(?s)` Equivalent

```python
text = "abc\ndef"
print(re.search("abc.def", text))            # None
print(re.search("(?s)abc.def", text))        # Match: abc\ndef
```

`(?s)` at the start of the pattern is the same as passing `re.DOTALL`.

---

#### 3. Using with `re.findall()`

```python
text = """<tag>
content line 1
content line 2
</tag>"""

# Match full tag content
pattern = re.compile(r"<tag>.*?</tag>", re.DOTALL)
print(pattern.findall(text))
# Output: ['<tag>\ncontent line 1\ncontent line 2\n</tag>']
```

---

### 🚫 Without `re.DOTALL`, You’d Have to Use Workarounds

```python
# Workaround (not ideal):
pattern = re.compile("Start[\s\S]*End")  # Match any whitespace or non-whitespace character
```

But `re.DOTALL` is cleaner and more readable.

---

### 📌 Summary

| Without `re.DOTALL`            | With `re.DOTALL`                     |
| ------------------------------ | ------------------------------------ |
| `.` does not match newline     | `.` matches newline (`\n`) too       |
| Useful for single-line parsing | Ideal for multi-line text extraction |
| May require regex tricks       | Simpler and readable                 |
