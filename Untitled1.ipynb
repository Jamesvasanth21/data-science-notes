```python
import cadquery as cq

# Parameters
diameter = 16  # mm
pitch = 2      # mm
length = 50    # mm
head_diameter = 24  # mm
head_height = 10    # mm

# Create screw shaft
shaft = cq.Workplane("XY").circle(diameter / 2).extrude(length)

# Create hex head
head = (
    cq.Workplane("XY")
    .polygon(6, head_diameter)
    .extrude(head_height)
    .translate((0, 0, length))
)

# Combine shaft and head
screw = shaft.union(head)

# Export to STEP
cq.exporters.export(screw, 'm16_screw.step')
```

```python
from cadquery import Workplane, exporters

# Shaft parameters
shaft_diameter = 36  # mm
shaft_length = 150   # mm

# Thread parameters
thread_major_diameter = 16  # mm
thread_pitch = 2            # mm
thread_length = 50          # mm

# Head parameters
head_width_across_flats = 24  # mm
head_height = 10              # mm

# Create shaft
shaft = Workplane("XY").circle(shaft_diameter / 2).extrude(shaft_length)

# Create threaded portion (modeled as a cylinder for simplicity)
thread = (
    Workplane("XY")
    .circle(thread_major_diameter / 2)
    .extrude(thread_length)
)

# Position threaded portion at the end of the shaft
thread = thread.translate((0, 0, shaft_length - thread_length))

# Create hex head
head = (
    Workplane("XY")
    .polygon(6, head_width_across_flats)
    .extrude(head_height)
    .translate((0, 0, shaft_length))
)

# Combine all parts
screw = shaft.union(thread).union(head)

# Export STEP file
step_file = "shaft_with_thread_and_head.step"
exporters.export(screw, step_file)

print(f"STEP file saved as {step_file}")
```
