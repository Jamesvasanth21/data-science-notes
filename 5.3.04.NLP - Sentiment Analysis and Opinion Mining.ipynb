### 📈 Sentiment Analysis and Opinion Mining

Sentiment analysis, also known as **opinion mining**, is a natural language processing (NLP) technique used to determine the emotional tone or opinion expressed in a piece of text. It's widely applied to analyze social media, customer reviews, and surveys to understand public sentiment toward a product, service, topic, or person.

-----

### 📚 Understanding Sentiment Analysis

Sentiment analysis works by classifying the polarity of a text, which can be **positive, negative, or neutral**. More advanced systems can detect finer emotions like anger, joy, sadness, or sarcasm. The core idea is to automate the process of understanding opinions, which is a major challenge for computers due to the subtleties of human language.

  * **Rule-Based Approach:** This method uses a predefined lexicon (a list of words and their associated sentiment scores). For example, "happy" and "good" might have a positive score, while "bad" and "terrible" have a negative score. The overall sentiment of a document is calculated by summing the scores of the words it contains.
  * **Machine Learning Approach:** This is the most common method today. It involves training a model on a large dataset of text that has been manually labeled with sentiment (e.g., positive, negative). The model learns the patterns and features that distinguish one sentiment from another.
  * **Hybrid Approach:** It is the combination of two or more approaches i.e. rule-based and Machine Learning approaches. The surplus is that the accuracy is high compared to the other two approaches.
----


### 🛠️ Feature Extraction for Sentiment Analysis

Feature extraction is the process of converting raw text into numerical features that a machine learning model can understand. Common techniques include:

  * **Bag-of-Words (BoW):** This is a simple but effective method. It represents text as a bag (multiset) of words, disregarding grammar and even word order, but keeping word frequency. For example, the sentence "This movie is great" would be represented by a vector counting the occurrences of "this," "movie," "is," and "great."
  * **Term Frequency-Inverse Document Frequency (TF-IDF):** This method goes a step further than BoW. It not only considers how often a word appears in a document but also how common it is across all documents. Words that are frequent in a specific document but rare in the overall corpus are given a higher weight, as they are likely more important for distinguishing that document.
  * **Word Embeddings:** This is a more advanced technique that represents words as dense vectors in a continuous vector space. Words with similar meanings are located closer to each other in this space. Popular examples include **Word2Vec** and **GloVe**. These methods capture semantic and syntactic relationships between words, which is a significant improvement over BoW or TF-IDF.

### **Different types of sentiment analysis**, with explanations to help you understand the distinctions and applications:

---

### ✅ **1. Fine-grained Sentiment Analysis**

* **Description**: Instead of just positive/negative, it classifies sentiment into multiple levels.
* **Typical labels**:
  • Very Positive
  • Positive
  • Neutral
  • Negative
  • Very Negative
* **Use cases**:
  • Product reviews where nuanced feedback matters
  • Social media monitoring with mixed sentiments

---

### ✅ **2. Binary Sentiment Analysis**

* **Description**: Classifies text into two categories — Positive or Negative.
* **Use cases**:
  • Simple customer feedback
  • Brand monitoring

---

### ✅ **3. Multiclass Sentiment Analysis**

* **Description**: Classifies text into more than two categories, like positive, negative, neutral, and mixed sentiments.
* **Use cases**:
  • Survey analysis
  • Complex conversational systems

---

### ✅ **4. Aspect-Based Sentiment Analysis (ABSA)**

* **Description**: Analyzes sentiments regarding specific aspects or attributes mentioned in the text.
* **Example**:
  “The camera is great, but the battery life is poor.”
  → Positive sentiment for “camera”, negative for “battery life”.
* **Use cases**:
  • E-commerce reviews
  • Customer support feedback

---

### ✅ **5. Emotion Detection**

* **Description**: Goes beyond polarity (positive/negative) and classifies sentiments into emotions like joy, sadness, anger, fear, surprise, disgust, etc.
* **Use cases**:
  • Mental health analysis
  • Social media emotion tracking

---

### ✅ **6. Intent-based Sentiment Analysis**

* **Description**: Detects underlying user intentions along with sentiment, like complaint, request, praise, or query.
* **Use cases**:
  • Chatbots
  • Customer service prioritization

---

### ✅ **7. Comparative Sentiment Analysis**

* **Description**: Identifies sentiment when users compare two or more products or services.
* **Example**:
  “Product A is better than Product B.”
  → Positive for Product A, negative for Product B
* **Use cases**:
  • Competitive analysis
  • Marketing strategies

---

### ✅ **8. Document-level Sentiment Analysis**

* **Description**: Determines the overall sentiment of a full document or article.
* **Use cases**:
  • News articles
  • Blogs

---

### ✅ **9. Sentence-level Sentiment Analysis**

* **Description**: Analyzes sentiment at the sentence level for more granular insights.
* **Use cases**:
  • Reviews
  • Feedback forms

---

### ✅ **10. Phrase-level Sentiment Analysis**

* **Description**: Focuses on individual phrases or expressions to understand subtle opinions.
* **Use cases**:
  • Social media hashtags
  • Micro-blogging platforms

---

### ✅ **11. Targeted Sentiment Analysis**

* **Description**: Focuses on a particular entity or topic mentioned within a text.
* **Example**:
  “I love the service, but the food is disappointing.”
  → Positive sentiment toward “service”, negative toward “food”.

### 💻 Building Sentiment Analysis Models

Here's a basic Python example using the **scikit-learn** library to build a sentiment analysis model. We'll use a simple dataset and the **TF-IDF** vectorizer with a **Multinomial Naive Bayes** classifier, which is a common choice for text classification tasks.

#### 1\. Setup and Data Preparation

First, you need to install the necessary libraries:

Next, let's create a small sample dataset. In a real-world scenario, you'd load a much larger dataset from a file.

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Sample data: a small dataset of reviews
data = {
    'text': ['I love this product!',
             'This is a terrible movie.',
             'The service was okay.',
             'Highly recommend.',
             'Absolutely awful.',
             'Not a fan of this.',
             'It was a good experience.',
             'The best purchase I have made so far.'
             ],
        'sentiment': ['positive',
                      'negative',
                      'neutral',
                      'positive',
                      'negative',
                      'negative',
                      'positive',
                      'positive'
                      ]
    }

df = pd.DataFrame(data)
```

#### 2\. Splitting the Data and Vectorizing

We need to split our data into training and testing sets. We then use the `TfidfVectorizer` to convert the text into numerical features.


```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['sentiment'], test_size=0.2, random_state=42)

# Create a TF-IDF Vectorizer
vectorizer = TfidfVectorizer()

# Fit the vectorizer on the training data and transform it
X_train_vec = vectorizer.fit_transform(X_train)

# Transform the test data using the *same* vectorizer
X_test_vec = vectorizer.transform(X_test)
```

```python
X_train_vec[0]
```

#### 3\. Training and Evaluating the Model

Now we can train a `MultinomialNB` model and evaluate its performance.

```python
y_pred[0]
```

```python
# Create and train a Multinomial Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train_vec, y_train)

# Predict the sentiment of the test data
y_pred = model.predict(X_test_vec)

# Evaluate the model
print(f"Accuracy: {accuracy_score(y_test, y_pred):.2f}")
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

#### 4\. Making a Prediction

Finally, let's see how the model performs on a new, unseen sentence.

```python
# Function to predict sentiment for new text
def predict_sentiment(text):
    text_vec = vectorizer.transform([text])
    prediction = model.predict(text_vec)
    return prediction[0]

# Example prediction
new_text = "I am so happy with the results!"
sentiment = predict_sentiment(new_text)
print(f'\nThe sentiment of "{new_text}" is: {sentiment}')
```
