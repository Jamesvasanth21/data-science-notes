# 🧪 Python Exception Handling
## Goal: Understand and Implement Exception Handling in Python Programs

**Topics Covered:** try, except, else, finally, raising exceptions, custom exceptions, best practices

## Introduction

Welcome to the exciting world of Python programming! As you write more complex programs, you'll inevitably encounter situations where things don't go exactly as planned. These "unplanned situations" can cause your program to crash, which isn't ideal for a good user experience. This is where **Exception Handling** comes in!

Exception handling is a powerful mechanism that allows you to gracefully manage errors and unexpected events in your code, preventing your program from crashing and making it more robust and user-friendly.

## 1. What is an Exception?

In Python, an **exception** is an event that occurs during the execution of a program that disrupts the normal flow of instructions. When an error occurs during runtime, Python creates an `Exception` object. If this object is not handled, it will cause the program to terminate abruptly and display an error message (a "traceback").

Think of it like driving a car:
* **Normal flow:** Driving smoothly on the road.
* **Exception:** Suddenly, a tire goes flat! This is an unexpected event.
* **Without handling:** If you don't know how to change a tire, your journey stops there.
* **With handling:** If you have a spare tire and know how to change it, you can fix the issue and continue your journey.

### Examples of common exceptions:

* **`ZeroDivisionError`**: Trying to divide a number by zero.
* **`NameError`**: Trying to use a variable that hasn't been defined.
* **`TypeError`**: Performing an operation on incompatible data types (e.g., adding a number to a string).
* **`IndexError`**: Trying to access an index that is out of range for a list or string.
* **`FileNotFoundError`**: Trying to open a file that doesn't exist.

Let's see some examples that *cause* exceptions:

```

# result = 10 / 0
# print(result) # This line will not be executed
# print("This variable does not exist!")
# print("This variable does not exist!")
# print("This variable does not exist!")
```

```
# Example 1: ZeroDivisionError
print("--- ZeroDivisionError ---")
try:
    result = 10 / 0
    print(result) # This line will not be executed
except ZeroDivisionError as e:
    print(f"Error: {e}")
    print("Cannot divide by zero!")

# Example 2: NameError
print("\n--- NameError ---")
try:
    print(undefined_variable)
except NameError as e:
    print(f"Error: {e}")
    print("This variable does not exist!")

# Example 3: TypeError
print("\n--- TypeError ---")
try:
    sum_value = 5 + "hello"
    print(sum_value)
except TypeError as e:
    print(f"Error: {e}")
    print("Cannot add a number and a string!")

# Example 4: IndexError
print("\n--- IndexError ---")
my_list = [1, 2, 3]
try:
    print(my_list[5])
except IndexError as e:
    print(f"Error: {e}")
    print("Index is out of range!")

# Example 5: FileNotFoundError
print("\n--- FileNotFoundError ---")
try:
    with open("non_existent_file.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError as e:
    print(f"Error: {e}")
    print("The file was not found!")
```

## 2. Handling Exceptions: The `try...except` Block

The core of exception handling in Python is the `try...except` block.

* The code that might raise an exception is placed inside the `try` block.
* If an exception occurs within the `try` block, the execution immediately jumps to the corresponding `except` block.
* The `except` block specifies the type of exception it wants to catch and contains the code to handle that specific exception.

### Basic `try...except` syntax:

```
# try:
#     # Code that might raise an exception
#     # ...
# except ExceptionType:
#     # Code to handle the specific ExceptionType
#     # ...
```

### Example: Handling `ZeroDivisionError`

```
print("--- Handling ZeroDivisionError ---")
try:
    numerator = 10
    denominator = int(input("Enter a denominator: ")) # User might enter 0
    result = numerator / denominator
    print(f"The result is: {result}")
except ZeroDivisionError:
    print("Error: You cannot divide by zero!")
    # try:
    #   print(f"The result is: {res}")
    # except IndexError:
    #   print("Error: Not defined")
```

### Catching Multiple Exceptions

You can have multiple `except` blocks to handle different types of exceptions. Python will execute the first `except` block that matches the type of exception raised.

```
print("\n--- Catching Multiple Exceptions ---")
try:
    num1 = int(input("Enter the first number: "))
    num2 = int(input("Enter the second number: "))
    result = num1 / num2
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except ValueError:
    print("Error: Invalid input. Please enter numbers only.")
```

### Catching Exceptions with a General `except` (Not Recommended for Best Practice)

You can use a general `except` block to catch any exception. However, it's generally **not recommended** to use a bare `except` statement as it can hide unexpected errors, making debugging difficult. It's better to be specific about the exceptions you expect.

```
print("\n--- General Except (Caution!) ---")
try:
    value = int("abc") # This will cause a ValueError
    print(10 / 0)        # This will cause a ZeroDivisionError
except Exception as e:
    print("An unknown error occurred!", e)
```

**Why is a general `except` block not recommended?**
It catches *all* exceptions, even those you didn't anticipate. This can mask bugs in your code, making it harder to debug because you don't know *what* went wrong. It's like having a universal "error" message for every problem.

### Getting the Exception Object and Information

You can capture the exception object itself using `as` to get more details about the error. This is very useful for debugging and providing more informative messages to the user.

```
print("\n--- Getting Exception Information ---")
try:
    my_list = [1, 2, 3]
    print(my_list[10]) # IndexError
except IndexError as e:
    print(f"An IndexError occurred: {e}")
    print(f"Type of error: {type(e)}")

try:
    age = int("twenty") # ValueError
except ValueError as err:
    print(f"A ValueError occurred: {err}")
    print(f"Error details: {err.args}") # Common way to get argument of exception
```

## 3. The `try...except...else` Block

The `else` block in a `try...except` statement is executed **only if no exception occurs** in the `try` block. It's a great place for code that should run only when everything in the `try` block is successful.

```
print("\n--- try...except...else ---")

def divide_numbers(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Error: Cannot divide by zero in the division function.")
    except TypeError:
        print("Error: Invalid types provided for division.")
    else:
        print(f"Division successful! Result: {result}")
        return result
    print("Operation failed.")
    return None

divide_numbers(10, 2)
divide_numbers(10, 0)
divide_numbers(10, "hello")
```

**When to use `else`?**
Use `else` for code that logically belongs to the "successful execution" path of the `try` block, but you want to ensure it only runs if no exception occurred. This helps keep the `try` block focused on the code that might *raise* exceptions.

## 4. The `try...finally` Block

The `finally` block is guaranteed to execute, **regardless of whether an exception occurred or not**. This makes it ideal for cleanup actions, such as closing files, releasing resources, or ensuring a specific state.

```
print("\n--- try...finally ---")

def process_file(filename):
    file_object = None
    try:
        print(f"Attempting to open {filename}...")
        file_object = open(filename, "r")
        content = file_object.read()
        print(f"File content:\n{content}")
        # Introduce an error to demonstrate finally
        # print(10 / 0)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An unexpected error occurred: {e}")
    finally:
        if file_object:
            file_object.close()
            print(f"File '{filename}' has been closed.")
        else:
            print(f"File '{filename}' was not opened or already closed.")

# Create a dummy file for demonstration
with open("my_data.txt", "w") as f:
    f.write("Line 1\nLine 2\nLine 3")

process_file("my_data.txt")
print("\n--- After first file processing ---")
process_file("non_existent_file.txt")
print("\n--- After second file processing ---")

# Let's see finally block with an unhandled exception (it will still execute)
try:
    print("Inside try block with a deliberate error...")
    result = 10 / 0 # This will raise ZeroDivisionError
except ZeroDivisionError: # This won't catch ZeroDivisionError
    print("Caught ValueError (but it's not the right one!)")
finally:
    print("This 'finally' block *always* executes, even if no matching except is found!")
print("Program continues after the finally block (if exception is handled or not re-raised).")
```

**When to use `finally`?**
Use `finally` for code that absolutely *must* be executed, regardless of the outcome of the `try` block. This is crucial for resource management.

## 5. Exception Arguments

As you saw earlier, when an exception occurs, it can carry an argument (or arguments) that provide details about the error. You can access these arguments using the `as` keyword in the `except` block.

Most standard exceptions have `args` as a tuple containing the error message.

```
print("\n--- Exception Arguments ---")
try:
    my_dict = {'a': 1, 'b': 2}
    print(my_dict['c']) # KeyError
except KeyError as e:
    print(f"KeyError occurred: {e}")
    print(f"Arguments of the exception: {e.args}")

try:
    import non_existent_module # ModuleNotFoundError
except ImportError as e:
    print(f"ImportError occurred: {e}")
    print(f"Arguments of the exception: {e.args}")
```

## 6. Standard Python Exceptions

Python has a rich hierarchy of built-in exceptions. Understanding these helps you write more specific and effective exception handlers. All built-in exceptions inherit from `BaseException`. Most common ones inherit from `Exception`.

Here are some common built-in exceptions you'll encounter:

* **`ArithmeticError`**: Base class for all errors that occur for numeric calculations.
    * `ZeroDivisionError`
    * `OverflowError`
* **`AttributeError`**: Raised when an attribute reference or assignment fails.
* **`EOFError`**: Raised when the `input()` function hits an end-of-file condition (EOF) without reading any data.
* **`ImportError`**: Raised when an `import` statement fails to find the module definition or when a `from ... import` fails to find a name that is to be imported.
    * `ModuleNotFoundError`
* **`IndexError`**: Raised when a sequence subscript is out of range.
* **`KeyError`**: Raised when a dictionary (mapping) key is not found.
* **`NameError`**: Raised when a local or global name is not found.
* **`OSError`**: Base class for I/O related errors.
    * `FileNotFoundError`
    * `PermissionError`
* **`RuntimeError`**: Raised when an error does not fall into any other category.
* **`TypeError`**: Raised when an operation or function is applied to an object of inappropriate type.
* **`ValueError`**: Raised when an operation or function receives an argument that has the right type but an inappropriate value.

You can view the full list in the official Python documentation.

## 7. Raising Exceptions (`raise`)

Sometimes, you might want to force an exception to occur if a certain condition is not met in your code. This is done using the `raise` statement. You can raise built-in exceptions or your own user-defined exceptions.

### Syntax:

```
raise ExceptionType("Optional error message")
```

### Example: Raising a `ValueError`

```
print("\n--- Raising Exceptions ---")

def validate_age(age):
    if not isinstance(age, (int, float)):
        raise TypeError("Age must be a number.")
    if age < 0:
        raise ValueError("Age cannot be negative.")
    elif age > 120:
        raise ValueError("Age seems unrealistically high.")
    else:
        print(f"Age {age} is valid.")

try:
    validate_age(30)
    validate_age(-5) # This will raise ValueError
except (TypeError, ValueError) as e:
    print(f"Validation Error: {e}")

try:
    validate_age(200) # This will raise ValueError
except ValueError as e:
    print(f"Another Validation Error: {e}")

try:
    validate_age("abc") # This will raise TypeError
except TypeError as e:
    print(f"Type Error: {e}")
```

### Re-raising an Exception

You can also re-raise an exception within an `except` block. This is useful if you want to perform some cleanup or logging, but then allow the exception to propagate up the call stack to be handled by a higher-level `except` block (or terminate the program).

```
print("\n--- Re-raising Exceptions ---")

def process_data(data):
    try:
        # if not data:
        #     raise ValueError("Data cannot be empty.")
        print(f"Processing data: {data}")
        # Simulate another potential error
        result = 10 / data
        print(f"Result: {result}")
    except ValueError as e:
        print(f"Logging error in process_data: {e}")
        # Re-raise the exception to be handled elsewhere
        raise
    except TypeError as e:
        print(f"Logging TypeError in process_data: {e}")
        raise
    except ZeroDivisionError as e:
        print(f"Logging ZeroDivisionError in process_data: {e}")
        raise

try:
    process_data(0) # This will cause ZeroDivisionError, which will be logged and re-raised
except ZeroDivisionError as e:
    print(f"Caught re-raised ZeroDivisionError in main program: {e}")

# try:
#     process_data(None) # This will cause ValueError, which will be logged and re-raised
# except ValueError as e:
#     print(f"Caught re-raised ValueError in main program: {e}")

# print("\n--- After re-raising examples ---")
```

## 8. User-defined Exceptions

While Python provides many built-in exceptions, sometimes you need to create your own custom exceptions to represent specific error conditions in your application. This makes your code more readable and robust, allowing you to catch application-specific errors precisely.

To create a user-defined exception, you simply create a new class that inherits from the `Exception` class (or one of its subclasses).

### Example: Creating a Custom Exception

Let's create a custom exception for a bank account system.

```
print("\n--- User-defined Exceptions ---")

class InsufficientFundsError(Exception):
    """Raised when an attempted withdrawal exceeds the available balance."""
    def __init__(self, message="Insufficient funds in the account.", balance=0, withdrawal_amount=0):
        self.message = message
        self.balance = balance
        self.withdrawal_amount = withdrawal_amount
        super().__init__(self.message) # Call the base class constructor

class BankAccount:
    def __init__(self, initial_balance):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        self.balance = initial_balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self.balance:
            raise InsufficientFundsError(
                f"Attempted to withdraw {amount}, but only {self.balance} available.",
                balance=self.balance, withdrawal_amount=amount
            )
        self.balance -= amount
        print(f"Withdrew {amount}. New balance: {self.balance}")

    def get_balance(self):
        return self.balance

# Demonstrate using the custom exception
account = BankAccount(100)

try:
    account.withdraw(50)
    account.deposit(200)
    account.withdraw(300) # This will raise InsufficientFundsError
except InsufficientFundsError as e:
    print(f"\nCaught custom exception: {e.message}")
    print(f"Current Balance: {e.balance}")
    print(f"Attempted Withdrawal: {e.withdrawal_amount}")
    print(f"Args: {e.args}")
except ValueError as e:
    print(f"\nCaught ValueError: {e}")
except Exception as e:
    print(f"\nCaught an unexpected error: {e}")

print(f"\nFinal balance: {account.get_balance()}")

try:
    account.withdraw(-10) # This will raise ValueError
except ValueError as e:
    print(f"\nCaught ValueError for invalid withdrawal: {e}")

try:
    BankAccount(-50) # This will raise ValueError
except ValueError as e:
    print(f"\nCaught ValueError for invalid initial balance: {e}")
```

**Why use custom exceptions?**

* **Clarity:** They clearly indicate what went wrong in your specific application context.
* **Specificity:** Allows you to catch and handle very particular error conditions, separating them from more general programming errors.
* **Maintainability:** Makes your code easier to understand and maintain.
* **Hierarchy:** You can build a hierarchy of custom exceptions, just like Python's built-in ones.

## Conclusion

Exception handling is an essential part of writing robust, reliable, and user-friendly Python programs. By using `try`, `except`, `else`, and `finally` blocks, and by understanding how to raise and define your own exceptions, you can anticipate and gracefully manage errors, preventing your programs from crashing and providing a better experience for your users.

**Key takeaways:**

* **Exceptions** are events that disrupt normal program flow.
* `try...except` blocks catch and handle exceptions.
* `else` runs if no exception occurs in `try`.
* `finally` always runs, regardless of exceptions, for cleanup.
* `raise` is used to trigger exceptions.
* **User-defined exceptions** enhance code clarity and error handling.
