## **2. Introduction to Image Formation**
Image formation explains how light from the environment becomes a digital image.

### **Key components**
- **Scene** – Objects that reflect/emit light.
- **Illumination** – Light interacting with objects.
- **Camera system** – Lens + sensor capturing light.

### **Image formation process**
1. Light hits an object.
2. Object reflects light based on surface properties.
3. Lens focuses light onto the sensor.
4. Sensor converts photons to electrical signals.
5. Signals are digitized to form an image.

### **Important concepts**
- **Projection** – 3D world mapped onto 2D plane (pinhole camera model).
- **Focal Length** – Controls zoom/magnification.
- **Depth of Field** – Region in focus.
- **Exposure** – Controlled by shutter speed, aperture, ISO.

## **3. Capture and Representation**

### **Image Capture**
- Modern cameras use **CMOS/CCD sensors**.
- Each pixel measures:
  - **Intensity** (brightness)
  - **Color** (via Bayer filter: R/G/B)
- Sensor data → digitized using an **ADC**.

### **Image Representation**
Digital images are stored as matrices.

- **Grayscale**:  
  A 2D matrix of pixel intensities (0–255 or normalized 0–1)

- **RGB Color Image**:  
  A 3D tensor: `Height × Width × 3`

- **Other formats**:
  - HSV (Hue, Saturation, Value)
  - YCbCr (Luma + chroma)
  - Depth images
  - Multi-spectral / hyper-spectral images

### **Spatial & Intensity Properties**
- **Resolution** – Number of pixels.
- **Bit-Depth** – Bits per pixel (8-bit, 16-bit).
- **Dynamic Range** – Ability to capture dark and bright details simultaneously.

## **4. Linear Filtering**
Linear filtering modifies an image by applying a filter kernel over each local region.

### **Purpose**
- Smoothing/noise reduction  
- Sharpening  
- Edge detection  
- Feature enhancement  

### **Definition**
A filter produces a weighted sum of neighboring pixels:


$ g(x,y) = \sum_{i=-k}^{k}\sum_{j=-k}^{k} h(i,j)\cdot f(x-i, y-j) $

Where:  
- `f` = input image  
- `h` = filter/kernel  
- `g` = output image  

### **Common linear filters**
- **Averaging filter** – blur  
- **Gaussian filter** – smooth  
- **Sobel filter** – edge detection  
- **Laplacian** – second derivative  

---

## **5. Correlation**
Correlation measures similarity between a kernel and a region of the image.

### **Operation**
- Slide kernel across the image.
- Multiply and sum overlapping values.
- Output represents similarity.

$ g(x,y) = \sum_{i=-k}^{k}\sum_{j=-k}^{k} h(i,j) \cdot f(x+i, y+j) $

### **Use cases**
- Template matching  
- Feature extraction  
- Blurring/sharpening  

### **Important note**
- **Correlation does NOT flip the kernel**.

---

## **6. Convolution**
Convolution is similar to correlation but the kernel is flipped horizontally and vertically.

### **Formula**

$ g(x,y) = \sum_{i=-k}^{k}\sum_{j=-k}^{k} h(i,j)\cdot f(x-i, y-j) $

### **Why convolution matters in deep learning**
- Core operation in **Convolutional Neural Networks (CNNs)**.
- Provides **translation invariance**.
- Enables **local receptive fields** and **parameter sharing**.
- Reduces number of learnable parameters drastically.

### **Examples**
- **Edge detection** – Sobel/Prewitt  
- **Blurring** – Gaussian convolution  
- **Sharpening** – High-pass filters  

### **Convolution in CNNs**
- Filters are **learned**, not predefined.
- Stacking layers → hierarchical feature extraction (edges → textures → shapes → objects).
