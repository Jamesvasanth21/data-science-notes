## 🧐 CNNs for Recognition and Verification

**Recognition** identifies a face in an image from a database of known faces (a 1-to-many problem). **Verification** confirms if a face in an image belongs to a specific person (a 1-to-1 problem). Instead of a standard softmax classifier, these tasks use a **distance-based approach** to compare face embeddings.

### **Siamese Networks**

A **Siamese Network** consists of two identical sub-networks that share weights. Each sub-network processes a different input image to produce a feature vector (an embedding). The distance between these embeddings is then calculated. A small distance indicates the faces are the same, while a large distance indicates they are different.

### **Loss Functions**

Training these networks requires specialized loss functions that learn to maximize the distance between dissimilar pairs and minimize it for similar pairs.

* **Contrastive Loss:** This loss function is applied to pairs of images. It pulls embeddings of similar pairs closer and pushes dissimilar pairs further apart in the embedding space.
    $$L(D) = \frac{1}{2N}\sum_{i=1}^{N} \left(y_i D_{i}^{2} + (1-y_i) \max(0, m - D_{i})^{2}\right)$$
    where $D$ is the Euclidean distance between embeddings, $y$ is 1 for similar pairs and 0 for dissimilar, and $m$ is the margin.
* **Triplet Loss:** This is a more effective loss function that operates on triplets of images: an **anchor** image, a **positive** image (same identity as the anchor), and a **negative** image (different identity). The goal is to ensure the distance between the anchor and positive is smaller than the distance between the anchor and negative, by at least a margin $\alpha$.
    $$L(a, p, n) = \max(0, ||f(a) - f(p)||^{2} - ||f(a) - f(n)||^{2} + \alpha)$$
* **Ranking Loss:** A general term for any loss function that learns a ranking. Triplet and contrastive losses are types of ranking loss.

## 🎯 CNNs for Detection

**Object Detection** is the task of identifying and localizing objects within an image by drawing a bounding box around them and assigning a class label.

### **Background of Object Detection**

Early methods used sliding windows, but this was computationally expensive. Modern approaches are based on deep learning and fall into two main categories:

1.  **Two-stage detectors:** These first propose regions of interest and then classify and refine them (e.g., R-CNN, Faster R-CNN).
2.  **One-stage detectors:** These predict bounding boxes and classes directly in a single pass (e.g., YOLO, SSD).

### **R-CNN (Region-based CNN)**

* **Process:**
    1.  Uses **Selective Search** to propose thousands of region proposals.
    2.  Resizes each proposal to a fixed size and feeds it into a pre-trained CNN.
    3.  Uses a Support Vector Machine (SVM) to classify the object within the region.
    4.  Uses a bounding box regressor to refine the box location.
* **Drawback:** Very slow due to the repeated CNN forward passes for each region.

### **Fast R-CNN**

* **Improvements:**
    1.  Feeds the entire image into the CNN once.
    2.  Uses **RoI (Region of Interest) Pooling** to extract a fixed-size feature vector for each region proposal from the final feature map, avoiding redundant calculations.
* **Drawback:** Still relies on an external algorithm (Selective Search) for region proposals.

### **Faster R-CNN**

* **Key Innovation:** Introduces a **Region Proposal Network (RPN)** that is trained to propose regions directly from the CNN feature map. This makes the entire pipeline end-to-end and significantly faster.

### **YOLO (You Only Look Once)**

* **Process:** Divides the image into a grid and has each grid cell predict bounding boxes and class probabilities simultaneously.
* **Advantages:** Extremely fast because it predicts everything in a single pass.
* **Drawbacks:** Less accurate than two-stage methods, especially for small objects.

### **SSD (Single Shot Detector)**

* **Process:** Similar to YOLO, but uses **multi-scale feature maps** to detect objects of different sizes. Small objects are detected by fine-grained feature maps, while large objects are detected by coarse-grained maps.
* **Advantages:** Faster than two-stage methods and more accurate than YOLO.

### **RetinaNet**

* **Key Innovation:** Addresses the **class imbalance** problem (too many easy background examples) in one-stage detectors with a new loss function called **Focal Loss**. This loss down-weights the contribution of "easy" examples, forcing the network to focus on "hard" examples.

## ✂️ CNNs for Segmentation

**Image Segmentation** is the task of assigning a class label to **every pixel** in an image.

### **FCN (Fully Convolutional Network)**

* **Concept:** Replaces the final dense layers of a CNN with convolutional layers. This allows the network to produce a segmentation map (a per-pixel output) instead of a single classification.
* **Process:** Uses a series of convolutional and pooling layers to downsample the image, and then a series of convolutional and upsampling layers to generate the final segmentation map.

### **SegNet**

* **Concept:** An encoder-decoder architecture. The **encoder** part is a standard CNN that downsamples the input image. The **decoder** uses the stored max-pooling indices from the encoder to perform intelligent upsampling, resulting in better detail preservation.

### **U-Net**

* **Concept:** A popular FCN-based architecture, particularly for biomedical image segmentation. Its "U" shape comes from its symmetric encoder-decoder structure.
* **Key Feature:** Uses **skip connections** that concatenate feature maps from the encoder path to the corresponding upsampled feature maps in the decoder path. This allows the network to access fine-grained details from the early layers, crucial for accurate segmentation.

### **Mask R-CNN**

* **Concept:** Extends Faster R-CNN by adding a third branch to predict a segmentation mask for each object, in addition to the class and bounding box.
* **Process:** It uses **RoIAlign** to extract features for each proposal, which is a more precise version of RoIPooling, leading to better mask quality.
