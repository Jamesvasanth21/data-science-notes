# Derive Gradients 

Derive **gradients** of a linear function $y = mx + b$ with respect to $m$ and $b$, in the context of **loss minimization** (like in linear regression).

---

### Step 1: The function

$$
y = mx + b
$$

* $m$ = slope (weight)
* $b$ = intercept (bias)
* $x$ = input feature
* $y$ = prediction

---

### Step 2: Loss function

Suppose we’re doing supervised learning with true value $y_{\text{true}}$.
We define a **loss function** $L$. A common choice is the **Mean Squared Error (MSE):**

$$
L = \frac{1}{2}(y_{\text{pred}} - y_{\text{true}})^2
$$

We add the $\tfrac{1}{2}$ just to make derivatives cleaner.

Here:

$$
y_{\text{pred}} = mx + b
$$

So:

$$
L = \frac{1}{2} \big( (mx+b) - y_{\text{true}} \big)^2
$$

---

### Step 3: Derivative wrt $m$ ($\frac{\partial L}{\partial m}$)

Using chain rule:

$$
\frac{\partial L}{\partial m} = \frac{\partial L}{\partial y_{\text{pred}}} \cdot \frac{\partial y_{\text{pred}}}{\partial m}
$$

1.

$$
\frac{\partial L}{\partial y_{\text{pred}}} = (y_{\text{pred}} - y_{\text{true}})
$$

2.

$$
\frac{\partial y_{\text{pred}}}{\partial m} = \frac{\partial (mx+b)}{\partial m} = x
$$

So:

$$
\frac{\partial L}{\partial m} = (y_{\text{pred}} - y_{\text{true}}) \cdot x
$$

---

### Step 4: Derivative wrt $b$ ($\frac{\partial L}{\partial b}$)

Again, chain rule:

$$
\frac{\partial L}{\partial b} = \frac{\partial L}{\partial y_{\text{pred}}} \cdot \frac{\partial y_{\text{pred}}}{\partial b}
$$

1.

$$
\frac{\partial L}{\partial y_{\text{pred}}} = (y_{\text{pred}} - y_{\text{true}})
$$

2.

$$
\frac{\partial y_{\text{pred}}}{\partial b} = 1
$$

So:

$$
\frac{\partial L}{\partial b} = (y_{\text{pred}} - y_{\text{true}})
$$

---

### Step 5: Interpretation

* $\frac{\partial L}{\partial m} = (y_{\text{pred}} - y_{\text{true}})\cdot x$
  👉 The error scaled by the input $x$. If $x$ is large, mistakes affect the slope more strongly.

* $\frac{\partial L}{\partial b} = (y_{\text{pred}} - y_{\text{true}})$
  👉 The error itself directly adjusts the intercept.

---

### Step 6: Update rule (Gradient Descent)

If learning rate = $\alpha$:

$$
m_{\text{new}} = m - \alpha \cdot \frac{\partial L}{\partial m}
$$

$$
b_{\text{new}} = b - \alpha \cdot \frac{\partial L}{\partial b}
$$

---

🔑 Intuition:

* $dl/dm$ → "How much the slope $m$ should move to reduce the error."
* $dl/db$ → "How much the intercept $b$ should move to reduce the error."

---

## Setup

**Input (column vector):**

$$
x=\begin{bmatrix}x_1\\x_2\end{bmatrix}\in\mathbb{R}^{2\times 1}
$$

**Hidden layer weights & bias (2 hidden units):**

$$
W^{(1)}=
\begin{bmatrix}
w^{(1)}_{11} & w^{(1)}_{12}\\
w^{(1)}_{21} & w^{(1)}_{22}
\end{bmatrix}\in\mathbb{R}^{2\times 2},
\qquad
b^{(1)}=
\begin{bmatrix}
b^{(1)}_{1}\\
b^{(1)}_{2}
\end{bmatrix}\in\mathbb{R}^{2\times 1}
$$

**Output layer weights & bias (single output):**

$$
W^{(2)}=
\begin{bmatrix}
w^{(2)}_{1} & w^{(2)}_{2}
\end{bmatrix}\in\mathbb{R}^{1\times 2},
\qquad
b^{(2)}\in\mathbb{R}
$$

Let the hidden activation be a generic $g(\cdot)$ (e.g., $\tanh$ or sigmoid), and for binary classification the output uses sigmoid $\sigma(\cdot)$.

## Forward pass (with shapes)

$$
\begin{aligned}
z^{(1)} &= W^{(1)}x + b^{(1)} \quad &&\in\mathbb{R}^{2\times 1}\\
a^{(1)} &= g\!\left(z^{(1)}\right) \quad &&\in\mathbb{R}^{2\times 1}\\[4pt]
z^{(2)} &= W^{(2)}a^{(1)} + b^{(2)} \quad &&\in\mathbb{R}^{1\times 1}\\
\hat y &= \sigma\!\left(z^{(2)}\right) \quad &&\in(0,1)
\end{aligned}
$$

**Expanded scalars:**

$$
\begin{aligned}
z^{(1)}_1 &= w^{(1)}_{11}x_1 + w^{(1)}_{12}x_2 + b^{(1)}_1\\
z^{(1)}_2 &= w^{(1)}_{21}x_1 + w^{(1)}_{22}x_2 + b^{(1)}_2\\
a^{(1)}_i &= g\!\left(z^{(1)}_i\right),\; i\in\{1,2\}\\
z^{(2)} &= w^{(2)}_1 a^{(1)}_1 + w^{(2)}_2 a^{(1)}_2 + b^{(2)}\\
\hat y &= \sigma(z^{(2)})
\end{aligned}
$$

## Loss (binary cross-entropy)

For a single example with label $y\in\{0,1\}$:

$$
\mathcal L = -\left[y\log \hat y + (1-y)\log(1-\hat y)\right]
$$

## Backpropagation

For **sigmoid + BCE**, the output-layer error simplifies to:

$$
\delta^{(2)} \;\equiv\; \frac{\partial \mathcal L}{\partial z^{(2)}} \;=\; \hat y - y
$$

Hidden layer error (elementwise “⊙”):

$$
\delta^{(1)} \;\equiv\; \frac{\partial \mathcal L}{\partial z^{(1)}} \;=\; \left(W^{(2)}\right)^{\!\top}\delta^{(2)} \;\; \odot \;\; g'\!\left(z^{(1)}\right)
$$

* If $g=\tanh$: $g'(z^{(1)}) = 1 - (a^{(1)})^2$ (elementwise).
* If $g=\sigma$: $g'(z^{(1)}) = a^{(1)}\odot(1-a^{(1)})$.

**Gradients (matrix form):**

$$
\begin{aligned}
\frac{\partial \mathcal L}{\partial W^{(2)}} &= \delta^{(2)} \, (a^{(1)})^{\!\top} \quad &&\in\mathbb{R}^{1\times 2}\\
\frac{\partial \mathcal L}{\partial b^{(2)}} &= \delta^{(2)} \quad &&\in\mathbb{R}\\[4pt]
\frac{\partial \mathcal L}{\partial W^{(1)}} &= \delta^{(1)} \, x^{\!\top} \quad &&\in\mathbb{R}^{2\times 2}\\
\frac{\partial \mathcal L}{\partial b^{(1)}} &= \delta^{(1)} \quad &&\in\mathbb{R}^{2\times 1}
\end{aligned}
$$

**Expanded scalars for the hidden layer grads:**

First compute

$$
\delta^{(1)}_1 = g'(z^{(1)}_1)\, w^{(2)}_1\, \delta^{(2)},
\qquad
\delta^{(1)}_2 = g'(z^{(1)}_2)\, w^{(2)}_2\, \delta^{(2)}.
$$

Then

$$
\begin{aligned}
\frac{\partial \mathcal L}{\partial w^{(1)}_{11}} &= \delta^{(1)}_1 \, x_1,
&
\frac{\partial \mathcal L}{\partial w^{(1)}_{12}} &= \delta^{(1)}_1 \, x_2,
&
\frac{\partial \mathcal L}{\partial b^{(1)}_1} &= \delta^{(1)}_1,\\
\frac{\partial \mathcal L}{\partial w^{(1)}_{21}} &= \delta^{(1)}_2 \, x_1,
&
\frac{\partial \mathcal L}{\partial w^{(1)}_{22}} &= \delta^{(1)}_2 \, x_2,
&
\frac{\partial \mathcal L}{\partial b^{(1)}_2} &= \delta^{(1)}_2.
\end{aligned}
$$

**Expanded scalars for the output layer grads:**

$$
\frac{\partial \mathcal L}{\partial w^{(2)}_1} = \delta^{(2)} a^{(1)}_1,\quad
\frac{\partial \mathcal L}{\partial w^{(2)}_2} = \delta^{(2)} a^{(1)}_2,\quad
\frac{\partial \mathcal L}{\partial b^{(2)}} = \delta^{(2)}.
$$

## (Optional) Gradient descent update

With learning rate $\eta$:

$$
W^{(\ell)} \leftarrow W^{(\ell)} - \eta \, \frac{\partial \mathcal L}{\partial W^{(\ell)}},\qquad
b^{(\ell)} \leftarrow b^{(\ell)} - \eta \, \frac{\partial \mathcal L}{\partial b^{(\ell)}}
\quad (\ell=1,2).
$$

# Understand Nueral Network Matrix Numerics Calculation

Below examples, **forward + backprop + single SGD update** on your 2–2–1 network (tanh hidden, sigmoid output, binary cross-entropy). No regularization, single example (batch size 1).

## Setup (numbers)

* Input/label: $x=\begin{bmatrix}0.6\\-1.2\end{bmatrix},\; y=1$
* Learning rate: $\eta=0.1$

**Layer 1 (2 hidden units):**

$$
W^{(1)}=\begin{bmatrix}0.3 & -0.7\\ 0.5 & 0.1\end{bmatrix},\quad
b^{(1)}=\begin{bmatrix}0.2\\-0.3\end{bmatrix}
$$

Activation $g=\tanh$.

**Layer 2 (single output):**

$$
W^{(2)}=\begin{bmatrix}0.4 & -0.2\end{bmatrix},\quad b^{(2)}=0.1
$$

Activation $\sigma$ (sigmoid).

---

## Forward pass

$$
\begin{aligned}
z^{(1)}_1 &= 0.3(0.6) + (-0.7)(-1.2) + 0.2 = 0.18 + 0.84 + 0.2 = 1.22\\
z^{(1)}_2 &= 0.5(0.6) + 0.1(-1.2) - 0.3 = 0.3 - 0.12 - 0.3 = -0.12\\
a^{(1)}_1 &= \tanh(1.22)\approx 0.8396541757\\
a^{(1)}_2 &= \tanh(-0.12)\approx -0.1194272985\\[4pt]
z^{(2)} &= 0.4(0.8396541757) + (-0.2)(-0.1194272985) + 0.1\\
&\approx 0.3358616703 + 0.0238854597 + 0.1 = 0.45974713\\
\hat y &= \sigma(0.45974713)\approx 0.6129541866
\end{aligned}
$$

**Loss (BCE, single example):**

$$
\mathcal L = -\big[y\log\hat y + (1-y)\log(1-\hat y)\big]
= -\log(0.6129541866)\approx 0.4894650822
$$

---

## Backprop

Output error (sigmoid + BCE):

$$
\delta^{(2)}=\frac{\partial\mathcal L}{\partial z^{(2)}}=\hat y - y
= 0.6129541866 - 1 = -0.3870458134
$$

**Gradients at output layer:**

$$
\begin{aligned}
\frac{\partial\mathcal L}{\partial W^{(2)}} &= \delta^{(2)}(a^{(1)})^\top
= \begin{bmatrix}
-0.3870458134\cdot 0.8396541757 &
-0.3870458134\cdot (-0.1194272985)
\end{bmatrix}\\
&= \begin{bmatrix}-0.3249846334 & 0.0462238359\end{bmatrix}\\
\frac{\partial\mathcal L}{\partial b^{(2)}} &= \delta^{(2)} = -0.3870458134
\end{aligned}
$$

**Hidden layer deltas:**

$$
g'(z)=1-\tanh^2(z)\quad\Rightarrow\quad
\begin{aligned}
g'(1.22)&\approx 0.2949808653,\quad g'(-0.12)\approx 0.9857371204\\
\delta^{(1)} &= (W^{(2)})^\top\delta^{(2)} \odot g'(z^{(1)})\\
\delta^{(1)}_1 &= (0.4\cdot -0.3870458134)\cdot 0.2949808653
= -0.0456684436\\
\delta^{(1)}_2 &= (-0.2\cdot -0.3870458134)\cdot 0.9857371204
= 0.0763050851
\end{aligned}
$$

**Gradients at hidden layer:**

$$
\frac{\partial\mathcal L}{\partial W^{(1)}}=\delta^{(1)}x^\top,\quad
\frac{\partial\mathcal L}{\partial b^{(1)}}=\delta^{(1)}
$$

Expanded:

$$
\begin{aligned}
\frac{\partial\mathcal L}{\partial w^{(1)}_{11}}&=\delta^{(1)}_1 x_1=-0.0274010661,&
\frac{\partial\mathcal L}{\partial w^{(1)}_{12}}&=\delta^{(1)}_1 x_2=0.0548021323,&
\frac{\partial\mathcal L}{\partial b^{(1)}_1}&=-0.0456684436\\
\frac{\partial\mathcal L}{\partial w^{(1)}_{21}}&=\delta^{(1)}_2 x_1=0.0457830511,&
\frac{\partial\mathcal L}{\partial w^{(1)}_{22}}&=\delta^{(1)}_2 x_2=-0.0915661021,&
\frac{\partial\mathcal L}{\partial b^{(1)}_2}&=0.0763050851
\end{aligned}
$$

---

## One SGD update ($\eta=0.1$)

Update rule: $\theta \leftarrow \theta - \eta\,\partial\mathcal L/\partial\theta$.

**Output layer:**

$$
\begin{aligned}
W^{(2)}_{\text{new}} &= \begin{bmatrix}0.4 & -0.2\end{bmatrix}
- 0.1\!\cdot\!\begin{bmatrix}-0.3249846334 & 0.0462238359\end{bmatrix}
= \begin{bmatrix}0.4324984633 & -0.2046223836\end{bmatrix}\\
b^{(2)}_{\text{new}} &= 0.1 - 0.1(-0.3870458134) = 0.1387045813
\end{aligned}
$$

**Hidden layer:**

$$
\begin{aligned}
W^{(1)}_{\text{new}} &=
\begin{bmatrix}
0.3 & -0.7\\
0.5 & 0.1
\end{bmatrix}
- 0.1\!\cdot\!
\begin{bmatrix}
-0.0274010661 & 0.0548021323\\
0.0457830511 & -0.0915661021
\end{bmatrix}\\
&=
\begin{bmatrix}
0.3027401066 & -0.7054802132\\
0.4954216949 & 0.1091566102
\end{bmatrix}\\[6pt]
b^{(1)}_{\text{new}} &=
\begin{bmatrix}0.2\\-0.3\end{bmatrix}
- 0.1\!\cdot\!
\begin{bmatrix}-0.0456684436\\ 0.0763050851\end{bmatrix}
=
\begin{bmatrix}0.2045668444\\ -0.3076305085\end{bmatrix}
\end{aligned}
$$


That’s one complete iteration: initialize → forward → loss → backprop → parameter update.
