## 1. What is an Exception?
An exception is an error that occurs during the execution of a program. When an exception occurs, the normal flow of the program is interrupted.

```python
print("--- Example of an Exception ---")
try:
    result = 10 / 0
except ZeroDivisionError as e:
    print("Exception occurred:", e)
```

## 2. Handling an Exception
We can handle exceptions using try...except blocks.

```python
print("\n--- Handling an Exception ---")
try:
    number = int(input("Enter a number: "))
    print("You entered:", number)
except ValueError:
    print("That's not a valid number!")
```

## 3. try...except...else
The else block runs if no exceptions were raised.

```python
print("\n--- try...except...else Example ---")
try:
    num = int(input("Enter an integer: "))
except ValueError:
    print("Invalid input!")
else:
    print("You entered:", num)
```

## 4. try...finally clause
The finally block always runs, no matter what.

```python
print("\n--- try...finally Example ---")
try:
    print("Inside try block")
    x = 1 / 1
finally:
    print("This always executes")
```

## 5. Argument of an Exception
You can access the exception's arguments.

```python
print("\n--- Exception Arguments ---")
try:
    x = int("hello")
except ValueError as err:
    print("Exception argument:", err)
```

## 6. Python Standard Exceptions
Python has many built-in exceptions: ZeroDivisionError, ValueError, TypeError, etc.

```python
print("\n--- Standard Exception Example: TypeError ---")
try:
    x = "abc" + 5
except TypeError as e:
    print("Caught TypeError:", e)
```

## 7. Raising an Exception
You can raise exceptions manually using `raise`.

```python
print("\n--- Raising an Exception ---")
def check_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    print("Valid age:", age)

try:
    check_age(-1)
except ValueError as e:
    print("Exception raised:", e)
```

## 8. User-Defined Exceptions
You can define your own custom exception classes.

```python
print("\n--- User-Defined Exception ---")
class MyCustomError(Exception):
    pass

def test(x):
    if x == 0:
        raise MyCustomError("x cannot be zero!")
    return 10 / x

try:
    test(0)
except MyCustomError as e:
    print("Caught custom exception:", e)
```
