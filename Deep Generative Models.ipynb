# 🎨 Deep Generative Models: A Data Scientist's Guide

Generative models are a class of machine learning models that are trained to generate new data instances that are similar to their training data. They are a fascinating and rapidly evolving area of deep learning.

### **Discriminative AI vs Generative AI**

| **Aspect**           | **Discriminative AI**                                                         | **Generative AI**                                                |                                                                                                             |
| ---------------------- | ------------------------------------------------------------------------------- | ------------------------------------------------------------------ | ----------------------------------------------------------------------------------------------------------- |
| **Main Purpose**       | Learn the mapping from input → output                                           | Learn the distribution of data and generate new samples            |                                                                                                             |
| **Focus**              | Distinguishing, predicting, or estimating                                       | Synthesizing, simulating, or augmenting data                       |                                                                                                             |
| **What it models**     | Conditional probability P(output, input); learns how outputs depend on inputs | Joint probability P(input, output) or data distribution P(data); learns how data is generated or structured |
| **Tasks**              | Classification (categorical labels), regression (continuous values)             | Image generation, text generation, audio synthesis, 3D modeling    |                                                                                                             |
| **Examples of models** | Logistic regression, linear regression, SVM, decision trees, neural classifiers | GANs, VAEs, Diffusion models, Transformers (GPT, DALL·E)           |                                                                                                             |
| **Output type**        | Labels or numerical predictions                                                 | Synthetic content such as images, text, shapes, audio              |                                                                                                             |
| **Input type**         | Observations, features, context                                                 | Context, random noise, or partial data                             |                                                                                                             |
| **Training approach**  | Minimize prediction error, maximize classification accuracy                     | Learn data distribution, adversarial training, reconstruction      |                                                                                                             |
| **Use cases**          | Spam detection, price forecasting, medical diagnosis                            | Artwork creation, chatbots, virtual simulations, data augmentation |                                                                                                             |

---

### Final Notes:

* **Discriminative AI** models the **conditional probability**, meaning it focuses on how the output depends on given inputs.
* **Generative AI** models the **joint or full distribution**, meaning it learns how the data itself is structured and how to create realistic samples.



## Review of Popular Generative Models

### **1. Generative Adversarial Networks (GANs)**

**GANs** are an ingenious framework consisting of two neural networks, a **Generator** and a **Discriminator**, that compete in a zero-sum game.

  * **The Generator ($G$)** takes a random noise vector as input and tries to create realistic-looking data (e.g., images).
  * **The Discriminator ($D$)** is a classifier that tries to distinguish between real data from the training set and fake data generated by the generator.

The two models are trained simultaneously. The generator's goal is to "fool" the discriminator, while the discriminator's goal is to not be fooled. This adversarial process forces the generator to produce increasingly realistic and high-quality data.

**Pros:**

  * Produce remarkably sharp and realistic images.
  * State-of-the-art for many image synthesis tasks.

**Cons:**

  * Can be notoriously difficult and unstable to train.
  * Prone to **mode collapse**, where the generator learns to produce only a limited variety of outputs.



### **2. Variational Autoencoders (VAEs)**

**VAEs** take a different approach. They are a type of **probabilistic autoencoder** that learns a latent representation of the data. They are built on an encoder-decoder architecture but with a probabilistic twist.

  * **The Encoder** maps the input data to a **probabilistic distribution** in a latent space (typically a Gaussian distribution), not a single point.
  * **The Decoder** samples from this latent distribution and attempts to reconstruct the original data.

The training objective is to minimize a loss function composed of two parts: a **reconstruction loss** (how well the data is reconstructed) and a **regularization loss** (which forces the latent distribution to be a standard Gaussian). This structure ensures the latent space is smooth and continuous, allowing for meaningful interpolation and generation.

**Pros:**

  * **Stable training** and easy to implement.
  * The latent space is meaningful and allows for easy interpolation and manipulation.

**Cons:**

  * Generated outputs are often blurry compared to GANs.
  * They don't produce the same level of fine-grained detail.

## Other Generative Models

### **3. PixelRNNs (Autoregressive Models)**

**PixelRNNs** belong to the class of **autoregressive models**. They generate data one element at a time, predicting each element based on the elements that came before it. For an image, this means generating one pixel at a time, conditioned on all previously generated pixels.

The joint probability of the data is factorized into a product of conditional probabilities:
$$P(x) = \prod_{i=1}^{n} P(x_i | x_1, ..., x_{i-1})$$

A PixelRNN uses a recurrent neural network to model this sequential generation. The output of the RNN at step $i-1$ is the input for the next step, allowing it to "remember" the context of previous pixels.

**Drawback:** The sequential nature of generation makes them very slow for high-resolution images.

### **4. NADE (Neural Autoregressive Distribution Estimator)**

**NADE** is an early example of an autoregressive model that improves upon the sequential nature of PixelRNNs. It uses a single feed-forward neural network to predict the conditional probabilities for each variable.

Unlike PixelRNNs, which require a sequential process for training and inference, NADE's structure allows the conditionals for all variables to be computed in parallel, making training more efficient. It maintains the autoregressive property by carefully masking the network to ensure that the prediction for $x\_i$ only depends on $x\_1, ..., x\_{i-1}$.

**NADE's Key Idea:** A single hidden layer can be used to estimate all conditional probabilities by using different masks for each output.

### **5. Normalizing Flows**

**Normalizing Flows** are a class of generative models that explicitly build a complex probability distribution by transforming a simple, known one (e.g., a standard Gaussian distribution).

The core idea is to apply a series of **invertible transformations** to a base distribution. Each transformation is a diffeomorphism, which means it is a smooth, invertible mapping.

**The Probability Density Rule:**
$$p_x(x) = p_z(z) \left| \det \left( \frac{\partial f^{-1}}{\partial x} \right) \right| = p_z(z) \left| \det \left( \frac{\partial f}{\partial z} \right)^{-1} \right|$$

Since the transformations are invertible, we can calculate the exact likelihood of a new data point, which is a significant advantage over GANs.

**Pros:**

  * **Stable training** (no adversarial game).
  * **Exact likelihood** calculation.
  * The latent space is interpretable.

**Cons:**

  * Requires a carefully designed architecture of invertible transformations.
  * Can be computationally expensive.
