## Understanding and Using `range()`

```python
# Basic usage of range
print(list(range(5)))           # 0 to 4
print(list(range(1, 6)))        # 1 to 5
print(list(range(0, 10, 2)))    # 0, 2, 4, 6, 8
```

```python
# Using range in a loop
for i in range(5):
    print(f"Number: {i}")
```

## List Comprehensions

```python
# Create a list of squares
squares = [x*x for x in range(5)]
print(squares)
```

```python
# Adding condition: even numbers
evens = [x for x in range(10) if x % 2 == 0]
print(evens)
```

## Dict and Set Comprehensions

```python
# Dictionary comprehension
squares_dict = {x: x*x for x in range(5)}
print(squares_dict)
```

```python
# Set comprehension
unique_mods = {x % 3 for x in range(10)}
print(unique_mods)
```

## Lambda Expressions

```python
# Basic lambda function
add = lambda x, y: x + y
print(add(2, 3))
```

```python
# Using lambda with map and filter
nums = [1, 2, 3, 4]
squared = list(map(lambda x: x*x, nums))
evens = list(filter(lambda x: x % 2 == 0, nums))
print("Squared:", squared)
print("Evens:", evens)
```

## Generator Functions

### 🧠 What is a Generator?

A **generator** is a type of function in Python that **yields values one at a time**, instead of returning them all at once.

It's like a **lazy list** — it gives you items only when you ask for them, which saves memory.

---

### 🔁 Normal Function vs Generator

#### Regular function (`return`):

```python
def get_numbers():
    return [1, 2, 3]

print(get_numbers())
```

📦 Returns the whole list at once.

---

#### Generator function (`yield`):

```python
def get_numbers(x):
    yield x * x
    yield 2
    yield 3

for num in get_numbers():
    print(num)
```

🚰 Yields one number at a time — **on demand**.

---

### 🧪 Real-World Analogy

Imagine a pizza place:

* A normal function bakes and hands you the **entire pizza** (even if you only want a slice).
* A generator gives you **one slice at a time**, as you eat.

---

### 🛠 Example: Countdown 

```python
def countdown(n):
    while n > 0:
        yield n
        n -= 1

for num in countdown(3):
    print(num)
```

### 🚀 Generator Expression (shortcut)

Like list comprehensions, but with parentheses:

```python
# Using generator expression
gen = (x*x for x in range(5))
for val in gen:
    print(val)
```

### 💡 When to Use Generators

* When working with **large datasets**
* When you don’t need **all values at once**
* When you want to **save memory**

---
Example below shows large range without memory issues:

```python
import sys
large_gen = (x for x in range(1000000))
print("Memory used by generator:", sys.getsizeof(large_gen), "bytes")
```

### What is 'yield'?

**Definition:**
The `yield` keyword is used in a **generator function** to return a value **without ending the function**.

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1
```

📌 Each call to `yield` **pauses** the function and **remembers** its state — it resumes from the same point on the next call.

---

**Think of `yield` like this:**

* `return` gives back one value and ends the function.
* `yield` gives one value but **keeps the function alive** for the next value.

```python
# 1. Fibonacci generator
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

list(fibonacci(10))
```
