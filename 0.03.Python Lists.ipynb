## Introduction

Welcome to the world of Python **Lists**! In programming, we often need to store collections of data. Instead of creating a separate variable for each item, Python Lists allow you to store multiple items in a single, organized structure.

Lists are one of the most versatile and widely used data structures in Python. They are ordered, changeable, and allow duplicate members.

---

## 1. What are Lists?

A **list** is a built-in Python data type that can store an ordered collection of items.

* **Ordered**: The items in a list have a defined order, and that order will not change unless you modify the list.
* **Changeable (Mutable)**: You can change, add, and remove items after the list has been created.
* **Allows Duplicates**: Lists can have items with the same value.
* **Heterogeneous**: Lists can contain items of different data types (integers, floats, strings, booleans, even other lists!).

### Creating a List

Lists are created by placing all the items inside square brackets `[]`, separated by commas.

```
# An empty list
empty_list = []
print(f"Empty list: {empty_list}")
print(f"Type of empty_list: {type(empty_list)}")

# A list of integers
numbers = [1, 2, 3, 4, 5]
print(f"List of numbers: {numbers}")

# A list of strings
fruits = ["apple", "banana", "cherry"]
print(f"List of fruits: {fruits}")

# A list with mixed data types
mixed_list = [1, "hello", 3.14, True]
print(f"Mixed list: {mixed_list}")

# A list with duplicate values
duplicate_items = ["red", "blue", "red", "green"]
print(f"List with duplicates: {duplicate_items}")
```

## 2. Accessing List Items (Indexing)

List items are accessed using their **index**. The index is the position number of an item in the list. Python uses **zero-based indexing**, meaning the first item has an index of `0`, the second item has an index of `1`, and so on.

### Positive Indexing

```
Numbers = [15, 40, 50, 100, 115, 140]
print(Numbers[3])
```

```
fruits = ["apple", "banana", "cherry", "orange", "kiwi"]

print(f"Original list: {fruits}")

# Access the first item (index 0)
print(f"First fruit: {fruits[0]}")

# Access the third item (index 2)
print(f"Third fruit: {fruits[2]}")

# Access the last item (index 4)
print(f"Last fruit: {fruits[4]}")
```

### Negative Indexing

You can also use negative indices to access items from the end of the list.
* `-1` refers to the last item.
* `-2` refers to the second to last item, and so on.

```
fruits = ["apple", "banana", "cherry", "orange", "kiwi"]

print(f"Original list: {fruits}")

# Access the last item using negative index
print(f"Last fruit (negative index): {fruits[-1]}")

# Access the second to last item
print(f"Second to last fruit: {fruits[-2]}")
```

### Slicing Lists

Slicing allows you to get a sub-section (a "slice") of a list. You specify a `start` index and an `end` index (exclusive).

Syntax: `list[start:end]`

* `start`: (Optional) The index where the slice begins (inclusive). Default is 0.
* `end`: (Optional) The index where the slice ends (exclusive). Default is the end of the list.

```
numbers = [10, 20, 30, 40, 50, 60, 70, 80, 90]

print(f"Original list: {numbers}")

# Get items from index 2 up to (but not including) index 5
print(f"Slice from index 2 to 5: {numbers[2:5]}") # Output: [30, 40, 50]

# Get items from the beginning up to index 4
print(f"Slice from start to index 4: {numbers[:5]}") # Output: [10, 20, 30, 40, 50]

# Get items from index 6 to the end
print(f"Slice from index 6 to end: {numbers[6:]}") # Output: [70, 80, 90]

# Get a copy of the entire list
print(f"Copy of the entire list: {numbers[:]}") # Output: [10, 20, 30, 40, 50, 60, 70, 80, 90]

# Slicing with a step (optional)
# Syntax: list[start:end:step]
print(f"Slice with step 2 (every other item): {numbers[::2]}") # Output: [10, 30, 50, 70, 90]
print(f"Slice in reverse: {numbers[::-1]}") # Output: [90, 80, 70, 60, 50, 40, 30, 20, 10]
```

## 3. Changing List Items

Since lists are mutable, you can change the value of a specific item by referring to its index.

```
fruits = ["apple", "banana", "cherry"]
print(f"Original list: {fruits}")

# Change the second item
fruits[1] = "grape"
print(f"List after changing item at index 1: {fruits}")

# Change a range of items
my_colors = ["red", "green", "blue", "yellow", "purple"]
print(f"Original colors: {my_colors}")
my_colors[1:3] = ["orange", "pink"] # Replace 'green' and 'blue' with 'orange' and 'pink'
print(f"Colors after changing slice: {my_colors}")

# If you insert more items than you remove, the list will grow
my_colors = ["red", "green", "blue"]
print(f"Colors before inserting more: {my_colors}")
my_colors[1:2] = ["orange", "pink", "brown"] # Replaces 'green' with three new items
print(f"Colors after inserting more: {my_colors}")
```

## 4. Adding Items to a List

There are several ways to add new items to a list:

### `append()`: Add an item to the end of the list

```
fruits = ["apple", "banana", "cherry"]
print(f"Original list: {fruits}")

fruits.append("orange")
print(f"After appending 'orange': {fruits}")
```

### `insert()`: Add an item at a specified index

```
fruits = ["apple", "banana", "cherry"]
print(f"Original list: {fruits}")

fruits.insert(1, "grape") # Insert "grape" at index 1
print(f"After inserting 'grape' at index 1: {fruits}")

fruits.insert(0, "kiwi") # Insert "kiwi" at the beginning
print(f"After inserting 'kiwi' at index 0: {fruits}")
```

### `extend()`: Add elements from another iterable (like another list) to the end of the current list

```
fruits = ["apple", "banana"]
more_fruits = ["cherry", "orange", "kiwi"]
print(f"Fruits: {fruits}")
print(f"More Fruits: {more_fruits}")

fruits.extend(more_fruits)
print(f"After extending fruits with more_fruits: {fruits}")

# You can extend with any iterable (e.g., a tuple)
veg = ["carrot", "broccoli"]
new_veg = ("potato", "spinach")
veg.extend(new_veg)
print(f"Vegetables after extending with tuple: {veg}")
```

```
fruits = ["apple", "banana"]
more_fruits = ["cherry", "orange", "kiwi"]
print(f"Fruits: {fruits}")
print(f"More Fruits: {more_fruits}")

fruits.extend(more_fruits)
print(f"After extending fruits with more_fruits: {fruits}")

# You can extend with any iterable (e.g., a tuple)
veg = ["carrot", "broccoli"]
print(type(veg))
new_veg = ("potato", "spinach")
print(type(new_veg))
veg.extend(new_veg)
print(f"Vegetables after extending with tuple: {veg}")
print(type(veg))
```

## 5. Removing Items from a List

Just as you can add items, you can also remove them.

### `remove()`: Remove the first occurrence of a specified value

```
fruits = ["apple", "banana", "cherry", "banana"]
print(f"Original list: {fruits}")

fruits.remove("banana") # Removes the first "banana"
print(f"After removing 'banana': {fruits}")

# If the item is not found, it will raise a ValueError
try:
    fruits.remove("grape")
except ValueError as e:
    print(f"Error trying to remove 'grape': {e}")
```

### `pop()`: Remove the item at a specified index (or the last item if no index is specified)

`pop()` also returns the removed item, which can be useful.

```
fruits = ["apple", "banana", "cherry", "orange"]
print(f"Original list: {fruits}")

removed_item = fruits.pop(1) # Remove item at index 1 ("banana")
print(f"After popping item at index 1: {fruits}")
print(f"Removed item: {removed_item}")

last_item = fruits.pop() # Remove the last item ("orange" in this case)
print(f"After popping last item: {fruits}")
print(f"Removed last item: {last_item}")

# If the index is out of range, it will raise an IndexError
try:
    fruits.pop(10)
except IndexError as e:
    print(f"Error trying to pop index 10: {e}")
```

### `del`: Delete items by index or slice, or delete the entire list

`del` is a statement, not a list method.

```
my_list = ["a", "b", "c", "d", "e"]
print(f"Original list: {my_list}")

del my_list[2] # Delete item at index 2 ("c")
print(f"After deleting item at index 2: {my_list}")

new_list = [10, 20, 30, 40, 50]
del new_list[1:4] # Delete items from index 1 up to (not including) 4
print(f"After deleting slice: {new_list}")

# Delete the entire list object
numbers_to_delete = [1, 2, 3]
print(f"List before deletion: {numbers_to_delete}")
del numbers_to_delete
try:
    print(numbers_to_delete) # This will cause a NameError
except NameError as e:
    print(f"Error: {e}. The list 'numbers_to_delete' no longer exists.")
```

### `clear()`: Empty the list

`clear()` removes all items from the list, but the list object itself remains.

```
my_list = [1, 2, 3, 4, 5]
print(f"Original list: {my_list}")

my_list.clear()
print(f"After clearing the list: {my_list}")
```

## 6. List Methods (Useful Built-in Functions)

Python lists come with several useful built-in methods (functions that belong to the list object).

### `len()`: Get the number of items in a list

`len()` is a built-in function, not a list method, but it's crucial for lists.

```
fruits = ["apple", "banana", "cherry", "orange"]
print(f"Fruits list: {fruits}")
print(f"Number of fruits: {len(fruits)}")

empty_list = []
print(f"Length of empty list: {len(empty_list)}")
```

### `sort()`: Sort the list in ascending order (in-place)

```
numbers = [5, 2, 8, 1, 9]
print(f"Original numbers: {numbers}")
numbers.sort()
print(f"Sorted numbers (ascending): {numbers}")

fruits = ["cherry", "apple", "banana", "orange"]
print(f"Original fruits: {fruits}")
fruits.sort()
print(f"Sorted fruits (alphabetical): {fruits}")

# Sort in descending order
numbers.sort(reverse=True)
print(f"Sorted numbers (descending): {numbers}")
```

### `sorted()`: Returns a *new* sorted list (does not modify the original)

This is a built-in function, not a list method.

```
original_numbers = [5, 2, 8, 1, 9]
print(f"Original numbers: {original_numbers}")

new_sorted_numbers = sorted(original_numbers)
print(f"New sorted list: {new_sorted_numbers}")
print(f"Original numbers (unchanged): {original_numbers}") # Original list remains the same

# You can also sort in reverse
reverse_sorted_numbers = sorted(original_numbers, reverse=True)
print(f"New reverse sorted list: {reverse_sorted_numbers}")
```

### `reverse()`: Reverse the order of elements (in-place)

```
my_list = ["a", "b", "c", "d"]
print(f"Original list: {my_list}")
my_list.reverse()
print(f"Reversed list: {my_list}")
```

### `count()`: Count the occurrences of a specific item

```
items = [1, 2, 2, 3, 1, 4, 2]
print(f"List: {items}")
print(f"Count of 1: {items.count(1)}")
print(f"Count of 2: {items.count(2)}")
print(f"Count of 5: {items.count(5)}")
```

### `index()`: Find the index of the first occurrence of a specified item

```
fruits = ["apple", "banana", "cherry", "orange"]
print(f"List: {fruits}")

print(f"Index of 'banana': {fruits.index('banana')}")

# If the item is not found, it raises a ValueError
try:
    print(f"Index of 'grape': {fruits.index('grape')}")
except ValueError as e:
    print(f"Error trying to find index of 'grape': {e}")
```

### `copy()`: Create a shallow copy of the list

Important for preventing unintended modifications to the original list when working with copies.

```
original_list = [1, 2, 3]
copied_list = original_list

copied_list[2] = 8

print(copied_list)
print(original_list)
```

```
original_list = [1, 2, 3]
copied_list = original_list.copy()

print(f"Original list: {original_list}")
print(f"Copied list: {copied_list}")

# Modify the copied list
copied_list.append(4)
print(f"Modified copied list: {copied_list}")
print(f"Original list (unchanged): {original_list}") # Original is not affected
```

**Note on `copy()` vs. assignment (`=`)**:
When you do `list2 = list1`, `list2` is *not* a new list; it's just another name (reference) pointing to the *same* list object in memory. Any changes to `list2` will also affect `list1`. `copy()` creates a truly separate list.

```
original_list = [1, 2, 3]
copied_list = original_list

copied_list[2] = 8

if copied_list == original_list:
  print(True)
else:
  print(False)
```

```
print(original_list)
print(copied_list)
```

```
original_list = [1, 2, 3]
copied_list = original_list.copy()

copied_list[2] = 8

if copied_list == original_list:
  print(True)
else:
  print(False)
```

```
list_a = [10, 20, 30]
list_b = list_a # This is a reference, not a copy!

list_b.append(40)

print(f"List A after modifying List B: {list_a}") # List A is also changed!
print(f"List B: {list_b}")
```

## 7. Iterating Through a List (Loops)

You can easily loop through the items in a list using a `for` loop.

```
fruits = ["apple", "banana", "cherry", "date"]

print("--- Iterating through fruits ---")
for fruit in fruits:
    print(fruit)

print("\n--- Iterating with index using enumerate() ---")
for index, fruit in enumerate(fruits):
    print(f"Fruit at index {index}: {fruit}")

print("\n--- Iterating with range(len()) ---")
for i in range(len(fruits)):
    print(f"Fruit at index {i}: {fruits[i]}")
```

## 8. List Comprehensions (Advanced, but Powerful!)

List comprehensions provide a concise way to create lists. They are often more readable and efficient than traditional `for` loops for list creation.

Syntax: `[expression for item in iterable if condition]`

* `expression`: The item you want to put into the new list.
* `item`: The variable representing each item in the iterable.
* `iterable`: The sequence you are iterating over (e.g., another list, a range).
* `condition`: (Optional) A filter that determines if the item should be included.

```
# Create a list of squares of numbers from 0 to 9
squares = [x**2 for x in range(10)]
print(f"Squares: {squares}")

# Create a list of even numbers from 0 to 19
even_numbers = [x for x in range(20) if x % 2 == 0]
print(f"Even numbers: {even_numbers}")

# Create a list of capitalized fruits
fruits = ["apple", "banana", "cherry"]
capitalized_fruits = [fruit.capitalize() for fruit in fruits]
print(f"Capitalized fruits: {capitalized_fruits}")

# Conditional expression in list comprehension
# new_list = [expression_if_true if condition else expression_if_false for item in iterable]
numbers = [1, 2, 3, 4, 5, 6]
odd_even_status = ["even" if num % 2 == 0 else "odd" for num in numbers]
print(f"Odd/Even status: {odd_even_status}")
```

## 9. Nested Lists (Lists of Lists)

Lists can contain other lists, creating nested structures. This is useful for representing matrices, grids, or hierarchical data.

```
# A simple 2x2 matrix
matrix = [
    [1, 2],
    [3, 4]
]
print(f"Matrix: {matrix}")

# Accessing elements in a nested list
# matrix[row_index][column_index]
print(f"First row, first element: {matrix[0][0]}") # Output: 1
print(f"Second row, second element: {matrix[1][1]}") # Output: 4

# A list of student records
students = [
    ["Alice", 20, "Computer Science"],
    ["Bob", 22, "Physics"],
    ["Charlie", 21, "Mathematics"]
]

print(f"Student records: {students}")
print(f"Bob's age: {students[1][1]}") # Output: 22

# Iterating through a nested list
print("\n--- Iterating through student records ---")
for student in students:
    name, age, major = student # Unpack the inner list
    print(f"Name: {name}, Age: {age}, Major: {major}")
```
