## Python Sets

````python
my_set = {1, 2, 3}
print(my_set)

# Add / Remove
my_set.add(4)
print(my_set)

my_set.update([5, 6])
print(my_set)

my_set.discard(2)
print(my_set)

# Set operations
set1 = {1, 2, 3}
set2 = {3, 4, 5}
print(set1.union(set2))
print(set1.intersection(set2))
print(set1.difference(set2))
````

## Binary Files and Pickle

````python
import pickle

data = {'id': 101, 'name': 'Alice'}

# Write binary with pickle
with open('data.pkl', 'wb') as f:
    pickle.dump(data, f)

# Read back
with open('data.pkl', 'rb') as f:
    loaded = pickle.load(f)
    print(loaded)
````

````python
# Write binary with pickle
with open('data.pkl', 'ab') as f:
    pickle.dump(data, f)

# Read back
with open('data.pkl', 'rb') as f:
     while True:
        try:
            obj = pickle.load(f)
            print(obj)
        except EOFError:
              break
````

## User Defined Functions

````python
def greet(name):
   return f"Hello, {name}!"

print(greet('Bob'))
````

````python
def greet_user(name, age: None):
    if age == None:
        return f"Hello, {name}! You are {age} years old."
    else:
       return f"Hello, {name}!"

print(greet_user('Bob'))
````

````python
from typing import Optional

def greet_user(name: str, age: Optional[int] = None):
    if age is not None:
        return f"Hello, {name}! You are {age} years old."
    else:
        return f"Hello, {name}!"

print(greet_user('Bob'))           # Hello, Bob!
print(greet_user('Alice', 30))     # Hello, Alice! You are 30 years old.
print(greet_user(30))
print(greet_user(name="James"))
````

````python
def greet_user(name, age=40):
    return f"Hello, {name}! You are {age} years old."

print(greet_user('Bob'))
print(greet_user('Alice', 50))
````

## Lambda (Anonymous Functions)

````python
square = lambda x: x * x
print(square(5))

nums = [1, 2, 3, 4]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
````

## Miscellneous

Here's an explanation of the code:

---

### 📌 What it does:

This code **saves data** to a binary file using `pickle` (serialization), and then **reads it back** (deserialization).

---

### 🔧 First Part: Write Binary with Pickle

```python
with open('data.pkl', 'ab') as f:
    pickle.dump(data, f)
```

* `open('data.pkl', 'ab')`:

  * `'a'` means *append* mode, so it doesn't overwrite existing content.
  * `'b'` stands for *binary* mode.
  * So `'ab'` means append binary data.

* `pickle.dump(data, f)`:

  * Serializes the `data` object and writes it to the file `f`.

This is useful when you want to **store multiple Python objects** (e.g., lists, dictionaries) one after another in a single file.

---

### 🔁 Second Part: Read Back

```python
with open('data.pkl', 'rb') as f:
    loaded = pickle.load(f)
    print(loaded)
```

* `open('data.pkl', 'rb')`:

  * Opens the file in *read binary* mode.

* `pickle.load(f)`:

  * Deserializes the first object from the file and loads it into memory.
  * If the file contains **multiple objects**, this will load only the first one. You’d need a loop to load all.

---

### ⚠️ Note:

If you've appended several objects with `dump()`, use a loop to read them:

```python
with open('data.pkl', 'rb') as f:
    while True:
        try:
            obj = pickle.load(f)
            print(obj)
        except EOFError:
            break  # End of file reached
```

---

### 🧠 Summary:

* `pickle.dump()` → write/append a Python object to a binary file.
* `pickle.load()` → read one Python object from a binary file.
* Use `'ab'` to avoid overwriting existing pickled data.
* Use a loop with `EOFError` to read multiple objects.
