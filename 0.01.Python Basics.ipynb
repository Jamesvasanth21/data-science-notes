##### Self-Refrential Structure
```python
alist = [1, "a", 8, 5.3]
alist.append(alist)
print(alist[4])
```

###### Step-by-step explanation:

1. Initial list:

   ```python
   alist = [1, "a", 8, 5.3]
   ```

2. After `alist.append(alist)`:

   * You're appending the **list itself** as the last element of the list.
   * So now `alist` becomes:

     ```python
     [1, "a", 8, 5.3, [...]]
     ```
   * More precisely:

     ```python
     alist = [1, "a", 8, 5.3, alist]
     ```

   This creates a **self-referential structure** — the last element is a reference to the entire list.

3. `print(alist[4])`:

   * `alist[4]` **is the list itself** (i.e., `alist`)
   * So the output will be:

     ```python
     [1, 'a', 8, 5.3, [...]]
     ```
   * In most Python interpreters, this prints something like:

     ```
     [1, 'a', 8, 5.3, [...]]
     ```

   This `[...]` notation is used to indicate **self-reference** — Python avoids infinite recursion in the printout.

---

###### Final Answer:

```
[1, 'a', 8, 5.3, [...]]
```

### Type Conversion (Casting) in Python

Python supports implicit and explicit type conversion. Explicit conversion (casting) is done using constructors like int(), float(), str(), bool(), list(), tuple(), set().

**Notes:**
- Converting float -> int truncates the fractional part (possible data loss).
- Converting non-numeric strings to int/float raises ValueError.
- Truthiness: bool(0) is False; any non-zero number is True.

````python
# String to int/float
s = "123"
print(s, '->', type(s))

i = int(s)    # '123' -> 123"
print(i, type(i))

f = float(s)  # '123' -> 123.0"
print(f, type(f))

# Float to int (truncates)",
x = 3.9
xi = int(x)   # 3
print(x, '-> int:', xi)

# Int/float to string
print(42, '-> str:', str(42))
print(3.14, '-> str:', str(3.14))

# Bool conversions
print('bool(0):', bool(0))
print('bool(1):', bool(1))
print('bool(""):', bool(""))
print('bool("Hello"):', bool("Hello"))
        
````

````python
# 1. Numeric Conversions
print('--- Numeric Conversions ---')
i = 10
f = 3.14
print('int -> float:', float(i))
print('float -> int (truncation):', int(f))
print('int(3.9):', int(3.9))
````

````python
# 2. String Conversions
print('\n--- String Conversions ---')
s_int = '123'
s_float = '45.67'
print("string '123' -> int:", int(s_int))
print("string '45.67' -> float:", float(s_float))
print("int 123 -> string:", str(123))
````

````python
# 3. Boolean Conversions
print('\n--- Boolean Conversions ---')
print('bool(1):', bool(1))
print('bool(0):', bool(0))
print('bool("hello"):', bool("hello"))
print('bool(""):', bool(""))
````

````python
# 4. Container Conversions
print('\n--- Container Conversions ---')
lst = [1, 2, 2, 3]
print('Original list:', lst)
print('list -> tuple:', tuple(lst))
print('list -> set (removes duplicates):', set(lst))
print('list -> dict (requires key-value pairs):', dict([('a', 1), ('b', 2)]))
````

````python
# 5. Invalid Conversions (Demonstrates Error Handling)
print('\n--- Invalid Numeric Conversion (Error Handling) ---')
# Demonstrates that converting an invalid string raises a ValueError
for val in ['10', '3.14', 'abc']:
    try:
        print(f"Attempting int('{val}'):", int(val))
    except ValueError as e:
        print(f"int('{val}') error:", e)
        
    try:
        print(f"Attempting float('{val}'):", float(val))
    except ValueError as e:
        print(f"float('{val}') error:", e)
````

````python
# 6. Bonus: String to list/tuple
print('\n--- String to Container Conversion ---')
hello = 'hello'
print('string -> list:', list(hello))
print('string -> tuple:', tuple(hello))
````

### Defining Classes in Python

````python
import time
"""
Record multiple running times.
"""
class Timer:
    def __init__(self):
        self.times = []

    def start(self):
        self.tik = time.time()

    def stop(self):
        self.times.append(time.time() - self.tik)
        return self.times[-1]
````

````python
timer = Timer()

timer.start()
for i in range(100):
    square(i)

print(timer.stop())

timer.start()
for i in range(10000):
    square(i)

print(timer.stop())

timer.start()
for i in range(1000000):
    square(i)

print(timer.stop())
````
